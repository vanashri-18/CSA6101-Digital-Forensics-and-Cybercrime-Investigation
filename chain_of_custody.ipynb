Aim:

To verify the Chain of Custody of digital evidence by generating and comparing SHA-256 hash values before and after modifying the evidence file.


Algorithm:


1.Start the program.

2.Create a sample digital evidence file.

3.Generate the SHA-256 hash value of the file.

4.Store the generated hash as the original evidence hash.

5.Generate the hash again without modifying the file.

6.Compare both hash values to verify file integrity.

7.Modify the contents of the evidence file.

8.Generate a new SHA-256 hash value.

9.Compare the new hash with the original hash.
10.If both hashes are identical, display "Chain of Custody Maintained."

11.Otherwise, display "Evidence Tampering Detected Chain of Custody Broken."

12.Stop the program.

```python
# ==========================================
# Chain of Custody Verification
# Using SHA-256 Hashing
# ==========================================

import hashlib

filename = "digital_evidence.txt"

# Step 1: Create digital evidence
with open(filename, "w") as file:
    file.write("This is confidential digital evidence.")

print("Evidence file created successfully.\n")

# Step 2: Function to calculate SHA-256 hash
def calculate_hash(file_name):

    sha256 = hashlib.sha256()

    with open(file_name, "rb") as file:
        while True:
            data = file.read(4096)

            if not data:
                break

            sha256.update(data)

    return sha256.hexdigest()

# Step 3: Original Hash
original_hash = calculate_hash(filename)

print("Original SHA-256 Hash:")
print(original_hash)
print()

# Step 4: Verify Integrity
verification_hash = calculate_hash(filename)

print("Verification SHA-256 Hash:")
print(verification_hash)
print()

if original_hash == verification_hash:
    print("Evidence Integrity Verified")
    print("Chain of Custody Maintained")
else:
    print("Evidence Tampered")
    print("Chain of Custody Broken")

print("\n--------------------------------------")

# Step 5: Simulate Tampering
with open(filename, "a") as file:
    file.write("\nUnauthorized Data Added")

print("\nEvidence has been modified.\n")

# Step 6: Recalculate Hash
new_hash = calculate_hash(filename)

print("New SHA-256 Hash:")
print(new_hash)
print()

# Step 7: Compare Hashes
if original_hash == new_hash:
    print("Chain of Custody Maintained")
else:
    print("Evidence Tampering Detected")
    print("Chain of Custody Broken")
```
