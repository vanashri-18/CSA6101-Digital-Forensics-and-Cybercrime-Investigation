**Aim**

To develop a Python hashing utility for forensic comparison of digital evidence by calculating MD5, SHA-1, and SHA-256 hashes for a file and comparing them with previously recorded hash values to verify whether the evidence remains unchanged.

**Algorithm**

Select a digital evidence file.

Open the file in binary mode.

Calculate its MD5 hash.

Calculate its SHA-1 hash.

Calculate its SHA-256 hash.

Display all calculated hash values.

Accept previously recorded hash values.

Compare each calculated hash with the corresponding recorded hash.

Report each algorithm as MATCH or NON-MATCH.

Interpret the results for forensic evidence verification.

```python
# ==============================================
# DIGITAL FORENSIC HASHING UTILITY
# ==============================================

import hashlib
from google.colab import files

print("=" * 70)
print("             DIGITAL EVIDENCE HASH VERIFICATION")
print("=" * 70)

# ------------------------------------------------
# 1. Upload Evidence File
# ------------------------------------------------

print("\nUpload the digital evidence file:")

uploaded = files.upload()

filename = list(uploaded.keys())[0]

# ------------------------------------------------
# 2. Calculate Hashes
# ------------------------------------------------

md5_hash = hashlib.md5()
sha1_hash = hashlib.sha1()
sha256_hash = hashlib.sha256()

with open(filename, "rb") as file:

    while True:

        data = file.read(4096)

        if not data:
            break

        md5_hash.update(data)
        sha1_hash.update(data)
        sha256_hash.update(data)

calculated_md5 = md5_hash.hexdigest()
calculated_sha1 = sha1_hash.hexdigest()
calculated_sha256 = sha256_hash.hexdigest()

# ------------------------------------------------
# 3. Display Calculated Hashes
# ------------------------------------------------

print("\n" + "-" * 70)
print("CALCULATED HASH VALUES")
print("-" * 70)

print("File     :", filename)
print("MD5      :", calculated_md5)
print("SHA-1    :", calculated_sha1)
print("SHA-256  :", calculated_sha256)

# ------------------------------------------------
# 4. Enter Previously Recorded Hashes
# ------------------------------------------------

print("\n" + "-" * 70)
print("ENTER PREVIOUSLY RECORDED HASH VALUES")
print("-" * 70)

recorded_md5 = input("Recorded MD5     : ").strip().lower()
recorded_sha1 = input("Recorded SHA-1   : ").strip().lower()
recorded_sha256 = input("Recorded SHA-256 : ").strip().lower()

# ------------------------------------------------
# 5. Compare Hashes
# ------------------------------------------------

md5_match = calculated_md5 == recorded_md5
sha1_match = calculated_sha1 == recorded_sha1
sha256_match = calculated_sha256 == recorded_sha256

# ------------------------------------------------
# 6. Display Verification Results
# ------------------------------------------------

print("\n" + "=" * 70)
print("                 HASH VERIFICATION RESULT")
print("=" * 70)

print(
    f"MD5      : {'MATCH' if md5_match else 'NON-MATCH'}"
)

print(
    f"SHA-1    : {'MATCH' if sha1_match else 'NON-MATCH'}"
)

print(
    f"SHA-256  : {'MATCH' if sha256_match else 'NON-MATCH'}"
)

# ------------------------------------------------
# 7. Overall Interpretation
# ------------------------------------------------

print("\n" + "-" * 70)
print("FORENSIC INTERPRETATION")
print("-" * 70)

if md5_match and sha1_match and sha256_match:

    print("All three hash values MATCH.")
    print("The file is consistent with the previously recorded")
    print("hash values and shows no detected change.")

else:

    print("One or more hash values do NOT MATCH.")
    print("The file may have been modified, replaced, or corrupted.")
    print("Further forensic examination is recommended.")

print("\nCryptographic hashes provide a digital fingerprint")
print("of evidence and help verify evidence integrity.")
print("SHA-256 is generally preferred for modern forensic")
print("verification because of its stronger security properties.")

print("\n" + "=" * 70)
```

**Result**

The Python hashing utility successfully calculated MD5, SHA-1, and SHA-256 hash values for the selected digital evidence and compared them with previously recorded values. Matching hashes indicate that the examined file is consistent with the recorded evidence, while a non-matching hash indicates that the file may have changed. Cryptographic hashing therefore provides an important mechanism for verifying the integrity and authenticity of digital forensic evidence.
