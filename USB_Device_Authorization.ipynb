**Aim**

To develop a Python program that analyzes a simulated Windows USB activity dataset and distinguishes authorized removable devices from unrecognized devices by comparing device identifiers with an investigator-maintained authorized-device list. The program reports the device identity, connection time, user, and authorization status.

**Algorithm**

Create or load the simulated USB activity dataset.

Create an authorized-device list containing trusted device identifiers.

Read each USB connection record.

Extract the device identifier, device name, connection time, and username.

Compare the device identifier with the authorized-device list.

Mark the device as AUTHORIZED if its identifier is present.

Mark it as UNRECOGNIZED if its identifier is absent.

Display the authorization status for every connection.

Extract all unrecognized USB connections.

Display a summary of unrecognized devices.

```python
# ==============================================
# USB DEVICE AUTHORIZATION ANALYZER
# ==============================================

import pandas as pd

# ------------------------------------------------
# 1. Simulated Windows USB Activity Dataset
# ------------------------------------------------

data = [
    ["2026-08-25 08:10:00", "USB-001",
     "SanDisk Cruzer", "Alice"],

    ["2026-08-25 08:25:00", "USB-002",
     "Kingston DataTraveler", "Bob"],

    ["2026-08-25 08:40:00", "USB-003",
     "Unknown USB Device", "Charlie"],

    ["2026-08-25 09:00:00", "USB-001",
     "SanDisk Cruzer", "Alice"],

    ["2026-08-25 09:20:00", "USB-004",
     "Generic Flash Drive", "David"],

    ["2026-08-25 09:45:00", "USB-002",
     "Kingston DataTraveler", "Bob"]
]

df = pd.DataFrame(
    data,
    columns=[
        "Connection_Time",
        "Device_ID",
        "Device_Name",
        "User"
    ]
)

# ------------------------------------------------
# 2. Investigator-Maintained Authorized List
# ------------------------------------------------

authorized_devices = {
    "USB-001",
    "USB-002"
}

print("=" * 85)
print("                 USB DEVICE AUTHORIZATION ANALYZER")
print("=" * 85)

# ------------------------------------------------
# 3. Determine Authorization Status
# ------------------------------------------------

df["Authorization_Status"] = df["Device_ID"].apply(
    lambda device:
        "AUTHORIZED"
        if device in authorized_devices
        else "UNRECOGNIZED"
)

# ------------------------------------------------
# 4. Display All Connections
# ------------------------------------------------

print("\n" + "-" * 85)
print("                    USB CONNECTION REPORT")
print("-" * 85)

print(
    df.to_string(index=False)
)

# ------------------------------------------------
# 5. Unrecognized USB Devices
# ------------------------------------------------

unrecognized = df[
    df["Authorization_Status"] == "UNRECOGNIZED"
]

print("\n" + "=" * 85)
print("                UNRECOGNIZED USB CONNECTIONS")
print("=" * 85)

if unrecognized.empty:

    print("No unrecognized USB devices detected.")

else:

    print(
        unrecognized[
            [
                "Connection_Time",
                "Device_ID",
                "Device_Name",
                "User",
                "Authorization_Status"
            ]
        ].to_string(index=False)
    )

# ------------------------------------------------
# 6. Summary
# ------------------------------------------------

print("\n" + "=" * 85)
print("                         SUMMARY")
print("=" * 85)

print("Total USB Connections :", len(df))

print(
    "Authorized Connections:",
    len(df[df["Authorization_Status"] == "AUTHORIZED"])
)

print(
    "Unrecognized Connections:",
    len(unrecognized)
)

print(
    "Unique Devices:",
    df["Device_ID"].nunique()
)

print("\nAuthorized Device IDs:")
for device in sorted(authorized_devices):
    print(" -", device)

print("\nUSB activity analysis completed.")
print("=" * 85)
```

**Result**

The Python program successfully analyzed the simulated Windows USB activity records and compared each device identifier against the investigator-maintained authorized-device list. It correctly classified connections as AUTHORIZED or UNRECOGNIZED and reported the device identity, connection time, user, and authorization status. The final summary identified 2 unrecognized USB connections, which can be prioritized for further forensic investigation.
