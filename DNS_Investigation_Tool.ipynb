```python
import socket

# Create domains.txt automatically
with open("domains.txt", "w") as file:
    file.write("google.com\n")
    file.write("youtube.com\n")
    file.write("invalid-domain-test.xyz\n")
    file.write("example.com\n")

# Dictionaries and lists
resolved_domains = {}
failed_domains = []
ip_domains = {}

# Read domains from text file
with open("domains.txt", "r") as file:
    domain_list = [line.strip() for line in file if line.strip()]

# Resolve each domain
for domain in domain_list:
    try:
        ip = socket.gethostbyname(domain)

        resolved_domains[domain] = ip

        if ip not in ip_domains:
            ip_domains[ip] = []

        ip_domains[ip].append(domain)

    except socket.gaierror:
        failed_domains.append(domain)

# Display results
print("DNS INVESTIGATION SUMMARY")
print("=" * 50)

print("\nRESOLVED DOMAINS")
for domain, ip in resolved_domains.items():
    print(domain, "->", ip)

print("\nDOMAINS SHARING SAME IP")
found_shared = False

for ip, domain_names in ip_domains.items():
    if len(domain_names) > 1:
        print(ip, "->", ", ".join(domain_names))
        found_shared = True

if not found_shared:
    print("No domains sharing the same IP.")

print("\nFAILED DNS RESOLUTION")
if failed_domains:
    for domain in failed_domains:
        print(domain)
else:
    print("No DNS resolution failures.")

print("\nSUMMARY")
print("Total Domains:", len(domain_list))
print("Successfully Resolved:", len(resolved_domains))
print("Resolution Failed:", len(failed_domains))
```
