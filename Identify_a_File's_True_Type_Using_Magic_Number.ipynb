**Aim**

To identify the true file type by examining its byte-level signature (magic number) using Python.

**Algorithm**

Import the required library (filetype).

Upload the file to be analyzed.

Read the file's byte-level signature (magic number).

Compare the signature with known file signatures using the filetype library.

Display the detected file extension and MIME type.

If no matching signature is found, display that the file type is unknown.

```python
# Install the required library
!pip install filetype

import filetype
from google.colab import files

# Upload file
print("Upload a file:")
uploaded = files.upload()

filename = list(uploaded.keys())[0]

# Identify file type using magic number
kind = filetype.guess(filename)

if kind is None:
    print("Unknown file type.")
else:
    print("File Name      :", filename)
    print("File Extension :", kind.extension)
    print("MIME Type      :", kind.mime)
```

**Result**

The Python program successfully identified the file's true type by analyzing its byte-level signature (magic number). The detected file extension and MIME type were displayed, even if the file extension had been altered.
