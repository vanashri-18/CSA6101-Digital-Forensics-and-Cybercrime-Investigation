**Aim**

To write a Python program that detects unauthorized modifications to a digital evidence file by comparing its current SHA-256 hash value with a previously stored baseline hash value.

**Algorithm**

Import the required Python libraries (hashlib and os).

Upload the digital evidence file.

Calculate the file's SHA-256 hash value.

If a baseline hash file does not exist, store the current hash as the baseline.

If a baseline hash file exists, read the stored baseline hash.


Compare the current hash with the baseline hash.

If both hashes match, report that the file is unchanged.

If the hashes differ, report that the file has been modified or tampered with.



```python
# Experiment 16
# Detect Unauthorized File Modifications using SHA-256 Hash

import hashlib
import os
from google.colab import files

# Upload evidence file
uploaded = files.upload()
filename = list(uploaded.keys())[0]

baseline_file = "baseline_hash.txt"

# Function to calculate SHA-256 hash
def calculate_sha256(file_path):
    sha256 = hashlib.sha256()

    with open(file_path, "rb") as file:
        while True:
            chunk = file.read(4096)
            if not chunk:
                break
            sha256.update(chunk)

    return sha256.hexdigest()

# Calculate current hash
current_hash = calculate_sha256(filename)

# Check for baseline hash
if not os.path.exists(baseline_file):
    with open(baseline_file, "w") as f:
        f.write(current_hash)

    print("Baseline hash created successfully.")
    print("SHA-256 Hash:", current_hash)

else:
    with open(baseline_file, "r") as f:
        stored_hash = f.read().strip()

    print("Stored Hash :", stored_hash)
    print("Current Hash:", current_hash)

    if stored_hash == current_hash:
        print("\nResult: File Integrity Verified.")
        print("No unauthorized modifications detected.")
    else:
        print("\nWARNING!")
        print("Unauthorized file modification detected!")
```

**Result**

The Python program successfully calculated the SHA-256 hash of the digital evidence file and compared it with a stored baseline hash.
