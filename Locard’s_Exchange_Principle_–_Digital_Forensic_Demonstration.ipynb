**Aim**

To demonstrate Locard’s Exchange Principle by recording a file’s creation, modification, and access timestamps before and after a controlled interaction, comparing the metadata, and identifying digital traces left by the interaction.

**Algorithm**

1. Create File
Create a test file containing sample forensic data.

2. Capture Metadata
Record the file’s creation, modification, and access timestamps along with the observation date and time.

3. Perform Operation
Perform a controlled operation such as opening and reading the test file.

4. Recapture Metadata
Record the file timestamps again after the controlled operation.

5. Compare Metadata
Compare the initial and final timestamps to identify any changes.

6. Identify Changes
Mark each timestamp as CHANGED or NO CHANGE.

7. Interpret Evidence
Separate the observed timestamp changes from the forensic interpretation of those changes.

```python
import os
import time
from datetime import datetime

FILE_NAME = "forensic_test_file.txt"


def get_time(timestamp):
    return datetime.fromtimestamp(timestamp).strftime(
        "%Y-%m-%d %H:%M:%S"
    )


def capture_metadata(filename):
    stat = os.stat(filename)

    return {
        "Creation Time": stat.st_ctime,
        "Modification Time": stat.st_mtime,
        "Access Time": stat.st_atime
    }


def display_metadata(title, metadata, observation_time):

    print("\n" + "=" * 65)
    print(title)
    print("=" * 65)

    print("Observation Date & Time:",
          observation_time.strftime("%Y-%m-%d %H:%M:%S"))

    for key, value in metadata.items():
        print(f"{key:<20}: {get_time(value)}")


# 1. Create test file
with open(FILE_NAME, "w") as file:
    file.write("Locard's Exchange Principle Demonstration\n")
    file.write("Digital forensic test evidence.\n")


# 2. Capture initial metadata
initial_observation = datetime.now()
initial_metadata = capture_metadata(FILE_NAME)

display_metadata(
    "INITIAL METADATA",
    initial_metadata,
    initial_observation
)


# 3. Controlled operation
print("\n" + "=" * 65)
print("CONTROLLED OPERATION")
print("=" * 65)

print("Operation: Opening and reading the test file.")

with open(FILE_NAME, "r") as file:
    content = file.read()

print("File accessed successfully.")

time.sleep(2)


# 4. Capture final metadata
final_observation = datetime.now()
final_metadata = capture_metadata(FILE_NAME)

display_metadata(
    "FINAL METADATA",
    final_metadata,
    final_observation
)


# 5. Compare metadata
print("\n" + "=" * 65)
print("METADATA COMPARISON")
print("=" * 65)

changes_found = False

for key in initial_metadata:

    before = initial_metadata[key]
    after = final_metadata[key]

    print(f"\n{key}")
    print("Before :", get_time(before))
    print("After  :", get_time(after))

    if before != after:
        print("Status : CHANGED")
        changes_found = True
    else:
        print("Status : NO CHANGE")


# 6. Observed facts
print("\n" + "=" * 65)
print("OBSERVED FACTS")
print("=" * 65)

print("1. Initial file metadata was captured.")
print("2. The file was opened and read.")
print("3. Final metadata was captured.")
print("4. Both metadata sets were compared.")

if initial_metadata["Access Time"] != final_metadata["Access Time"]:
    print("5. Access timestamp changed after file interaction.")
else:
    print("5. Access timestamp showed no observable change.")


# 7. Forensic interpretation
print("\n" + "=" * 65)
print("FORENSIC INTERPRETATION")
print("=" * 65)

if initial_metadata["Access Time"] != final_metadata["Access Time"]:
    print("The access-time change is consistent with")
    print("interaction with the digital evidence.")
else:
    print("No access-time change was observed.")
    print("This does not prove that the file was never accessed.")

print("\nLocard's Principle:")
print("Digital interaction may leave observable traces.")

print("\n" + "=" * 65)
print("DEMONSTRATION COMPLETE")
print("=" * 65)
```

**Result**

The Locard’s Exchange Principle was successfully demonstrated. The program captured and compared the file timestamps before and after a controlled operation. The observed change in the access timestamp provides a digital trace consistent with interaction, demonstrating how filesystem metadata can contribute to forensic examination.
