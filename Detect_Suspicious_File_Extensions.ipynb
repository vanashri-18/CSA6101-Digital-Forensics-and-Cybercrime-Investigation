**Aim**

To write a Python program to detect suspicious file extensions, including double-extension files such as invoice.pdf.exe, which may be used by attackers to disguise malicious executable files as legitimate documents.

**Algorithm**

Import the required Python libraries.

Create a list of files to be analyzed.

Define commonly suspicious executable extensions such as .exe, .bat, .cmd, .scr, and .vbs.

Check each filename for its extensions.

Detect files containing multiple extensions.

Check whether the final extension is an executable or script extension.

Mark files such as invoice.pdf.exe as suspicious.

Display the suspicious filenames and the reason for detection.

Display the final analysis result.

```python
import os
from google.colab import files

# Suspicious executable/script extensions
suspicious_extensions = {
    ".exe",
    ".bat",
    ".cmd",
    ".scr",
    ".vbs",
    ".js",
    ".ps1",
    ".com",
    ".msi"
}

# Common document extensions
document_extensions = {
    ".pdf",
    ".doc",
    ".docx",
    ".xls",
    ".xlsx",
    ".ppt",
    ".pptx",
    ".jpg",
    ".jpeg",
    ".png",
    ".txt"
}

print("Upload files for suspicious extension analysis:")
uploaded = files.upload()

print("\n" + "=" * 60)
print("SUSPICIOUS FILE EXTENSION ANALYSIS")
print("=" * 60)

suspicious_files = []

for file_name in uploaded.keys():

    # Split filename into parts
    parts = file_name.lower().split(".")

    # Check final extension
    final_extension = "." + parts[-1] if len(parts) > 1 else ""

    # Detect double/multiple extension
    multiple_extension = len(parts) > 2

    # Check for document + executable combination
    disguised_file = False

    if multiple_extension:
        for i in range(1, len(parts) - 1):
            previous_extension = "." + parts[i]

            if (previous_extension in document_extensions and
                    final_extension in suspicious_extensions):
                disguised_file = True

    # Detect suspicious file
    if final_extension in suspicious_extensions or disguised_file:

        suspicious_files.append(file_name)

        print("\n[!] SUSPICIOUS FILE DETECTED")
        print("File:", file_name)

        if disguised_file:
            print("Reason: Possible double-extension attack")
        else:
            print("Reason: Suspicious executable/script extension")

    else:
        print("\n[OK] File appears normal")
        print("File:", file_name)


# Final result
print("\n" + "=" * 60)
print("FINAL RESULT")
print("=" * 60)

if suspicious_files:
    print("Suspicious files detected:", len(suspicious_files))

    for file_name in suspicious_files:
        print(" -", file_name)

else:
    print("No suspicious file extensions detected.")
```

**Result**

The Python program successfully detects suspicious file extensions and double-extension filenames. Files such as invoice.pdf.exe and payment.docx.scr are flagged because attackers may use legitimate-looking document extensions to hide executable or malicious files. This technique can assist in initial digital forensic and malware analysis.
