**Aim**

To develop a Python program that identifies the actual format of a file by examining its magic number (file signature), compares it with the supplied file extension, and determines whether the extension matches the actual file format.

**Algorithm**

1. Select Files
Select files with known and deliberately incorrect extensions.

2. Read Header
Read the first few bytes of each file in binary mode.

3. Load Signatures
Maintain a collection of common magic numbers such as PDF, JPEG, PNG, and ZIP.

4. Compare Bytes
Compare the file's header bytes with the stored magic numbers.

5. Detect Format
Identify the actual file format based on the matching signature.

6. Check Extension
Extract the supplied file extension.

7. Compare Formats
Compare the detected format with the supplied extension.

8. Display Result
Report the file name, supplied extension, detected format, and whether they agree.

```python
import os
from google.colab import files

# ---------------------------------------------------------
# FILE SIGNATURE / MAGIC NUMBER ANALYZER
# ---------------------------------------------------------

print("=" * 70)
print("             FILE SIGNATURE ANALYZER")
print("=" * 70)

# ---------------------------------------------------------
# COMMON MAGIC NUMBERS
# ---------------------------------------------------------

MAGIC_NUMBERS = {

    # PDF
    b"%PDF": {
        "format": "PDF Document",
        "extension": ".pdf"
    },

    # JPEG
    b"\xFF\xD8\xFF": {
        "format": "JPEG Image",
        "extension": ".jpg / .jpeg"
    },

    # PNG
    b"\x89PNG\r\n\x1a\n": {
        "format": "PNG Image",
        "extension": ".png"
    },

    # GIF
    b"GIF87a": {
        "format": "GIF Image",
        "extension": ".gif"
    },

    b"GIF89a": {
        "format": "GIF Image",
        "extension": ".gif"
    },

    # ZIP
    b"PK\x03\x04": {
        "format": "ZIP Archive",
        "extension": ".zip"
    },

    # RAR
    b"Rar!\x1a\x07": {
        "format": "RAR Archive",
        "extension": ".rar"
    },

    # EXE / DLL
    b"MZ": {
        "format": "Windows Executable",
        "extension": ".exe / .dll"
    }
}


# ---------------------------------------------------------
# FORMAT DETECTION
# ---------------------------------------------------------

def detect_file_format(filename):

    with open(filename, "rb") as file:

        # Read first 16 bytes
        header = file.read(16)

    for signature, information in MAGIC_NUMBERS.items():

        if header.startswith(signature):

            return (
                information["format"],
                information["extension"],
                signature
            )

    return "Unknown Format", "Unknown", None


# ---------------------------------------------------------
# EXTENSION CHECK
# ---------------------------------------------------------

def check_extension(filename, expected_extension):

    supplied_extension = os.path.splitext(filename)[1].lower()

    expected_extensions = [
        ext.strip().lower()
        for ext in expected_extension.split("/")
    ]

    if supplied_extension in expected_extensions:

        return supplied_extension, "MATCH"

    else:

        return supplied_extension, "MISMATCH"


# ---------------------------------------------------------
# UPLOAD FILES
# ---------------------------------------------------------

print("\nUpload at least THREE files for testing.")

uploaded = files.upload()

if len(uploaded) < 3:

    print("\nWARNING:")
    print("Please upload at least three different file types.")

else:

    print("\nFiles uploaded successfully:", len(uploaded))


    # -----------------------------------------------------
    # ANALYZE EACH FILE
    # -----------------------------------------------------

    for filename in uploaded.keys():

        print("\n" + "=" * 70)
        print("FILE ANALYSIS")
        print("=" * 70)

        print("File Name:", filename)

        # Detect actual format
        detected_format, expected_extension, signature = (
            detect_file_format(filename)
        )

        # Get supplied extension
        supplied_extension = os.path.splitext(
            filename
        )[1].lower()

        print("\nSupplied Extension:")
        print(supplied_extension)

        print("\nMagic Number:")

        with open(filename, "rb") as file:
            header = file.read(16)

        print(header.hex(" "))

        print("\nDetected Format:")
        print(detected_format)

        # -------------------------------------------------
        # Compare extension
        # -------------------------------------------------

        if detected_format != "Unknown Format":

            extension_status = check_extension(
                filename,
                expected_extension
            )

            print("\nExtension Check:")
            print(extension_status[1])

            if extension_status[1] == "MATCH":

                print(
                    "\nConclusion: File extension agrees "
                    "with the detected format."
                )

            else:

                print(
                    "\nConclusion: File extension does NOT "
                    "agree with the detected format."
                )

                print(
                    "Possible renamed or misleading file."
                )

        else:

            print("\nExtension Check: CANNOT DETERMINE")
            print(
                "No matching magic number was found."
            )

        print("\n" + "=" * 70)


print("\nFILE SIGNATURE ANALYSIS COMPLETED")
```

**Result**

The program successfully identified the actual file format using magic numbers instead of relying on the filename extension. Three different file types were tested with deliberately incorrect extensions. The program detected the actual formats and correctly reported the extension mismatches, demonstrating how file signatures can be used in digital forensics to identify files whose extensions have been deliberately or accidentally changed.
