```python
# Digital Evidence Chain of Custody System

evidence = {}
custody_history = []

# Function to add evidence details
def add_evidence():
    evidence["id"] = input("Enter Evidence ID: ")
    evidence["source"] = input("Enter Evidence Source: ")
    evidence["date_time"] = input("Enter Acquisition Date and Time: ")
    evidence["investigator"] = input("Enter Investigator Name: ")
    evidence["storage"] = input("Enter Storage Location: ")
    evidence["hash"] = input("Enter SHA-256 Hash Value: ")

# Function to add custody transfer
def add_transfer():
    transfer = {
        "from": input("Transferred From: "),
        "to": input("Transferred To: "),
        "date_time": input("Transfer Date and Time: "),
        "purpose": input("Purpose of Transfer: "),
        "hash": input("SHA-256 Hash at Transfer: ")
    }

    custody_history.append(transfer)
    print("\nCustody transfer added successfully!\n")

# Function to display complete evidence history
def display_history():
    print("\n" + "=" * 55)
    print("DIGITAL EVIDENCE CHAIN OF CUSTODY")
    print("=" * 55)

    print("\nEVIDENCE DETAILS")
    print("Evidence ID:", evidence["id"])
    print("Source:", evidence["source"])
    print("Acquisition Date/Time:", evidence["date_time"])
    print("Investigator:", evidence["investigator"])
    print("Storage Location:", evidence["storage"])
    print("Original SHA-256:", evidence["hash"])

    print("\nCUSTODY HISTORY")

    for i, transfer in enumerate(custody_history, start=1):
        print(f"\nTransfer {i}")
        print("From:", transfer["from"])
        print("To:", transfer["to"])
        print("Date/Time:", transfer["date_time"])
        print("Purpose:", transfer["purpose"])

        # Verify integrity
        if transfer["hash"] == evidence["hash"]:
            print("Integrity Status: CONSISTENT")
        else:
            print("Integrity Status: HASH MISMATCH")

# Main program
add_evidence()

while True:
    choice = input("\n1. Add Transfer\n2. Display History\n3. Exit\nChoose: ")

    if choice == "1":
        add_transfer()

    elif choice == "2":
        display_history()

    elif choice == "3":
        print("Program Closed.")
        break

    else:
        print("Invalid Choice!")
```
