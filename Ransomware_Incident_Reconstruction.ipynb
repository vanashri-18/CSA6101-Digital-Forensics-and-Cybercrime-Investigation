**Aim**

To develop a Python program that reconstructs a simulated ransomware incident from chronological process, file, rename, and network events, identifies the earliest suspicious event, subsequent file activity, and final observed activity, and presents the evidence supporting each stage.

**Algorithm**

Create a chronological simulated ransomware incident log.

Read timestamp, event type, process, path, and description.

Sort all records chronologically.

Identify suspicious process execution events.

Identify subsequent file modification events.

Identify rapid file-renaming events and unusual extensions.

Identify network activity associated with the incident.

Determine the earliest suspicious event.

Determine the final observed event.

Divide the incident into stages.

Display the reconstructed sequence with supporting evidence.


```python
# ==============================================
# RANSOMWARE INCIDENT RECONSTRUCTION
# ==============================================

import pandas as pd

# ------------------------------------------------
# 1. Simulated Chronological Incident Log
# ------------------------------------------------

data = [
    ["2026-08-25 10:00:00", "PROCESS",
     "winword.exe", "C:/Program Files/Office/winword.exe",
     "Normal document process started"],

    ["2026-08-25 10:02:00", "NETWORK",
     "winword.exe", "192.168.1.50",
     "Outbound connection detected"],

    ["2026-08-25 10:04:00", "PROCESS",
     "powershell.exe",
     "C:/Windows/System32/WindowsPowerShell/powershell.exe",
     "PowerShell executed by user"],

    ["2026-08-25 10:06:00", "FILE_MODIFY",
     "powershell.exe",
     "C:/Finance/report1.docx",
     "Document modified"],

    ["2026-08-25 10:07:00", "FILE_MODIFY",
     "powershell.exe",
     "C:/Finance/report2.xlsx",
     "Spreadsheet modified"],

    ["2026-08-25 10:08:00", "FILE_RENAME",
     "powershell.exe",
     "C:/Finance/report1.docx -> report1.locked",
     "File extension changed to .locked"],

    ["2026-08-25 10:09:00", "FILE_RENAME",
     "powershell.exe",
     "C:/Finance/report2.xlsx -> report2.locked",
     "File extension changed to .locked"],

    ["2026-08-25 10:11:00", "FILE_MODIFY",
     "powershell.exe",
     "C:/HR/employee_data.pdf",
     "Document modified"],

    ["2026-08-25 10:12:00", "FILE_RENAME",
     "powershell.exe",
     "C:/HR/employee_data.pdf -> employee_data.locked",
     "File extension changed to .locked"],

    ["2026-08-25 10:15:00", "NETWORK",
     "powershell.exe",
     "203.0.113.20",
     "Final outbound network connection observed"]
]

df = pd.DataFrame(
    data,
    columns=[
        "Timestamp",
        "Event_Type",
        "Process",
        "Object",
        "Description"
    ]
)

# ------------------------------------------------
# 2. Normalize Timestamp
# ------------------------------------------------

df["Timestamp"] = pd.to_datetime(
    df["Timestamp"]
)

# ------------------------------------------------
# 3. Sort Chronologically
# ------------------------------------------------

df = df.sort_values(
    "Timestamp"
).reset_index(drop=True)

print("=" * 100)
print("                 RANSOMWARE INCIDENT RECONSTRUCTION")
print("=" * 100)

# ------------------------------------------------
# 4. Identify Suspicious Events
# ------------------------------------------------

suspicious_processes = [
    "powershell.exe",
    "cmd.exe",
    "wscript.exe",
    "cscript.exe"
]

suspicious = df[
    (
        df["Process"]
        .str.lower()
        .isin(suspicious_processes)
    )
    |
    (
        df["Event_Type"]
        .isin([
            "FILE_MODIFY",
            "FILE_RENAME"
        ])
    )
]

# ------------------------------------------------
# 5. Earliest Suspicious Event
# ------------------------------------------------

earliest = suspicious.iloc[0]

# ------------------------------------------------
# 6. File Activity
# ------------------------------------------------

file_activity = df[
    df["Event_Type"].isin([
        "FILE_MODIFY",
        "FILE_RENAME"
    ])
]

# ------------------------------------------------
# 7. Final Observed Activity
# ------------------------------------------------

final_event = df.iloc[-1]

# ------------------------------------------------
# 8. Reconstructed Timeline
# ------------------------------------------------

print("\n" + "=" * 100)
print("                    RECONSTRUCTED SEQUENCE")
print("=" * 100)

for i, row in df.iterrows():

    print(
        f"{i+1}. {row['Timestamp']} | "
        f"{row['Event_Type']} | "
        f"{row['Process']} | "
        f"{row['Object']}"
    )

# ------------------------------------------------
# 9. Stage 1
# ------------------------------------------------

print("\n" + "=" * 100)
print("STAGE 1 - EARLIEST SUSPICIOUS EVENT")
print("=" * 100)

print("Timestamp :", earliest["Timestamp"])
print("Event     :", earliest["Event_Type"])
print("Process   :", earliest["Process"])
print("Object    :", earliest["Object"])
print("Evidence  :", earliest["Description"])

# ------------------------------------------------
# 10. Stage 2
# ------------------------------------------------

print("\n" + "=" * 100)
print("STAGE 2 - SUBSEQUENT FILE ACTIVITY")
print("=" * 100)

if file_activity.empty:

    print("No file activity detected.")

else:

    print(
        file_activity[
            [
                "Timestamp",
                "Event_Type",
                "Process",
                "Object",
                "Description"
            ]
        ].to_string(index=False)
    )

# ------------------------------------------------
# 11. Stage 3
# ------------------------------------------------

print("\n" + "=" * 100)
print("STAGE 3 - FINAL OBSERVED ACTIVITY")
print("=" * 100)

print("Timestamp :", final_event["Timestamp"])
print("Event     :", final_event["Event_Type"])
print("Process   :", final_event["Process"])
print("Object    :", final_event["Object"])
print("Evidence  :", final_event["Description"])

# ------------------------------------------------
# 12. Evidence Summary
# ------------------------------------------------

print("\n" + "=" * 100)
print("                       EVIDENCE SUMMARY")
print("=" * 100)

print(
    "Earliest Suspicious Event :",
    earliest["Timestamp"],
    "-",
    earliest["Event_Type"]
)

print(
    "File Activity Events      :",
    len(file_activity)
)

print(
    "Final Observed Event      :",
    final_event["Timestamp"],
    "-",
    final_event["Event_Type"]
)

print("\nIncident reconstruction completed.")
print("=" * 100)
```

**Result**

The program successfully reconstructed the simulated incident chronologically. It identified PowerShell execution at 10:04 as the earliest suspicious event, followed by multiple file modifications and .locked extension changes. The activity then continued across another directory, and the 10:15 network connection was identified as the final observed activity. Each stage is supported by the corresponding log records, providing an evidence-based sequence for forensic investigation.
