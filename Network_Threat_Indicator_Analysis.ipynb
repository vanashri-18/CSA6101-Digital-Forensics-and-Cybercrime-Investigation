**Aim**

To develop a Python network-log analyzer that compares source IP addresses against a predefined set of threat indicators (IOCs) and retrieves the related connection records. The program summarizes the number of events, contacted destinations, ports, and timestamps associated with each flagged source address.

**Algorithm**

Create a simulated network connection log.

Define a predefined list of threat-indicator IP addresses.

Read each network record from the log.

Compare the source IP with the threat-indicator list.

Extract all matching network records.

Group matching records by source IP.

Count the number of connection events.

Identify contacted destination addresses and ports.

Record the timestamps associated with each flagged IP.

Display the detailed evidence and generate a summary.

```python
# ==============================================
# NETWORK THREAT INDICATOR ANALYZER
# ==============================================

import pandas as pd

# ------------------------------------------------
# 1. Simulated Network Log
# ------------------------------------------------

data = [
    ["2026-08-25 09:00:00", "192.168.1.10",
     "203.0.113.10", 80],

    ["2026-08-25 09:05:00", "10.0.0.5",
     "198.51.100.20", 443],

    ["2026-08-25 09:10:00", "192.168.1.10",
     "203.0.113.10", 443],

    ["2026-08-25 09:15:00", "10.0.0.8",
     "192.0.2.50", 22],

    ["2026-08-25 09:20:00", "10.0.0.5",
     "198.51.100.20", 8080],

    ["2026-08-25 09:25:00", "192.168.1.25",
     "203.0.113.50", 80],

    ["2026-08-25 09:30:00", "10.0.0.5",
     "203.0.113.99", 443],

    ["2026-08-25 09:35:00", "192.168.1.10",
     "198.51.100.40", 22]
]

df = pd.DataFrame(
    data,
    columns=[
        "Timestamp",
        "Source_IP",
        "Destination_IP",
        "Port"
    ]
)

df["Timestamp"] = pd.to_datetime(
    df["Timestamp"]
)

# ------------------------------------------------
# 2. Predefined Threat Indicators
# ------------------------------------------------

threat_indicators = {
    "10.0.0.5",
    "192.168.1.10"
}

print("=" * 95)
print("                 NETWORK THREAT INDICATOR ANALYZER")
print("=" * 95)

print("\nThreat Indicator IP Addresses:")

for ip in sorted(threat_indicators):
    print(" -", ip)

# ------------------------------------------------
# 3. Compare Source IPs with IOCs
# ------------------------------------------------

flagged = df[
    df["Source_IP"].isin(
        threat_indicators
    )
].copy()

# ------------------------------------------------
# 4. Display Matching Records
# ------------------------------------------------

print("\n" + "=" * 95)
print("                    FLAGGED CONNECTION RECORDS")
print("=" * 95)

if flagged.empty:

    print("No threat-indicator addresses found.")

else:

    print(
        flagged[
            [
                "Timestamp",
                "Source_IP",
                "Destination_IP",
                "Port"
            ]
        ].to_string(index=False)
    )

# ------------------------------------------------
# 5. IOC Summary
# ------------------------------------------------

print("\n" + "=" * 95)
print("                    IOC CONNECTION SUMMARY")
print("=" * 95)

for ip in sorted(threat_indicators):

    records = flagged[
        flagged["Source_IP"] == ip
    ]

    if records.empty:

        print(
            f"\n{ip} : No matching events"
        )

        continue

    print("\n" + "-" * 95)

    print(
        f"Threat Indicator: {ip}"
    )

    print(
        "Number of Events :",
        len(records)
    )

    print(
        "Destinations     :",
        ", ".join(
            sorted(
                records["Destination_IP"]
                .unique()
            )
        )
    )

    print(
        "Ports            :",
        ", ".join(
            str(x)
            for x in sorted(
                records["Port"].unique()
            )
        )
    )

    print(
        "First Timestamp  :",
        records["Timestamp"].min()
    )

    print(
        "Last Timestamp   :",
        records["Timestamp"].max()
    )

# ------------------------------------------------
# 6. Overall Summary
# ------------------------------------------------

print("\n" + "=" * 95)
print("                         FINAL SUMMARY")
print("=" * 95)

print(
    "Total Network Events :",
    len(df)
)

print(
    "Flagged Events       :",
    len(flagged)
)

print(
    "Flagged Source IPs   :",
    flagged["Source_IP"].nunique()
)

print(
    "Threat Indicators    :",
    len(threat_indicators)
)

print("\nNetwork IOC analysis completed.")
print("=" * 95)
```

**Result**

The Python program successfully compared network source addresses against a predefined threat-indicator list and retrieved all matching connection records. In the simulated dataset, 2 source IP addresses matched the indicators and generated 6 flagged events. For each flagged address, the program reported the number of events, destination IPs, destination ports, first timestamp, and last timestamp, providing the investigator with supporting network evidence for further analysis.
