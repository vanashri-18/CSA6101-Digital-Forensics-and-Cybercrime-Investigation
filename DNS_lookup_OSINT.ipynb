Aim:

To develop a Python program that performs a DNS lookup for a given domain name and retrieves its corresponding IP address as a basic OSINT (Open Source Intelligence) reconnaissance technique.

Algorithm

1.Start the program.

2.Import the socket module.

3.Accept the domain name from the user.

4.Use the gethostbyname() function to perform a DNS lookup.

5.Retrieve the IP address associated with the domain.

6.Display the domain name and its IP address.

7.If the domain is invalid, display an error message.

8.Stop the program.

```python
import socket

print("========== DNS LOOKUP TOOL ==========")

domain = input("Enter Domain Name (Example: google.com): ")

try:
    ip_address = socket.gethostbyname(domain)
    print("\n========== RESULT ==========")
    print("Domain Name :", domain)
    print("IP Address  :", ip_address)
except socket.gaierror:
    print("Invalid Domain Name or DNS Lookup Failed!")
```

Result:

The program executed successfully.
