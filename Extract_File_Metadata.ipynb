**Aim**

To write a Python program that extracts the metadata of a digital evidence file, including the file name, file size, creation time, modification time, and access time.

**Algorithm**

Import the required Python libraries (os, time, and google.colab.files).

Upload the digital evidence file.

Read the file information using the os module.

Extract the file name and file size.

Obtain the file creation time, last modification time, and last access time.

Convert the timestamps into a human-readable
format.

Display all the extracted metadata.

```python
# Experiment 17
# Extract File Metadata

import os
import time
from google.colab import files

# Upload file
uploaded = files.upload()

# Get uploaded filename
filename = list(uploaded.keys())[0]

# File metadata
file_size = os.path.getsize(filename)
creation_time = os.path.getctime(filename)
modification_time = os.path.getmtime(filename)
access_time = os.path.getatime(filename)

# Display metadata
print("="*50)
print("        FILE METADATA REPORT")
print("="*50)
print("File Name          :", filename)
print("File Size          :", file_size, "bytes")
print("Creation Time      :", time.ctime(creation_time))
print("Modification Time  :", time.ctime(modification_time))
print("Access Time        :", time.ctime(access_time))
print("="*50)
```

**Result**

The Python program successfully extracted the metadata of the uploaded digital evidence file, including the file name, file size, creation time, modification time, and access time.
