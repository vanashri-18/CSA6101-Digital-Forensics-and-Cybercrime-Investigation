**Aim**

To develop a Python program that automatically extracts URLs, email addresses, IPv4 addresses, and domain names from a text file, removes duplicate indicators, stores them separately, and displays the number of unique indicators found.

**Algorithm**

Read the recovered text file.

Use regular expressions to identify digital indicators.

Extract URLs, email addresses, IPv4 addresses, and domain names.

Remove duplicate values using sets.

Store each category separately.

Display the extracted indicators and their unique counts.

```python
import re

# Create a dummy evidence.txt file for demonstration
with open("evidence.txt", "w") as file:
    file.write("Visit example.com for more info. You can also reach us at info@example.org or sales@anothersite.net. \nOur IP address is 192.168.1.1. \nCheck out https://www.google.com and http://test.info/path?query=1. \nAnother IP: 10.0.0.5. \nMore domains: sub.domain.co.uk, www.microsoft.com. \nSome text with no indicators. example.com info@example.org")

# Read text file
with open("evidence.txt", "r") as file:
    text = file.read()

# Extract indicators
urls = set(re.findall(r'https?://[^\s]+', text))

emails = set(re.findall(
    r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}', text
))

ips = set(re.findall(
    r'\b(?:\d{1,3}\.){3}\d{1,3}\b', text
))

domains = set(re.findall(
    r'\b(?:[A-Za-z0-9-]+\.)+[A-Za-z]{2,}\b', text
))

# Display results
print("URLs:", urls)
print("Email Addresses:", emails)
print("IPv4 Addresses:", ips)
print("Domain Names:", domains)

print("\nUnique Indicator Count")
print("URLs:", len(urls))
print("Emails:", len(emails))
print("IPv4 Addresses:", len(ips))
print("Domains:", len(domains))
```

**Result**

The Python program successfully extracted URLs, email addresses, IPv4 addresses, and domain names from the recovered text file. Duplicate indicators were removed, each category was stored separately, and the number of unique indicators was displayed for further forensic analysis.
