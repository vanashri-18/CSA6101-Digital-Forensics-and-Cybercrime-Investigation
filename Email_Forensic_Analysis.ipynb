```python
# Email Forensic Analysis

emails = [
    {
        "sender": "support@trustedbank.com",
        "recipient": "user@example.com",
        "timestamp": "2026-08-29 09:00",
        "subject": "Account Statement",
        "attachment": "statement.pdf",
        "url": "https://trustedbank.com/login"
    },
    {
        "sender": "security@bank-alert.xyz",
        "recipient": "user@example.com",
        "timestamp": "2026-08-29 09:15",
        "subject": "URGENT: Verify Your Account",
        "attachment": "verify.exe",
        "url": "http://bank-alert.xyz/verify"
    },
    {
        "sender": "unknown@secure-check.net",
        "recipient": "user@example.com",
        "timestamp": "2026-08-29 09:30",
        "subject": "Password Reset Required",
        "attachment": "password.zip",
        "url": "http://secure-check.net/login"
    }
]

# Predefined investigation criteria
suspicious_domains = ["bank-alert.xyz", "secure-check.net"]
suspicious_words = ["urgent", "verify", "password", "reset"]
suspicious_extensions = [".exe", ".zip"]

results = []

for email in emails:
    score = 0
    reasons = []

    # Check sender domain
    domain = email["sender"].split("@")[1]

    if domain in suspicious_domains:
        score += 3
        reasons.append("Suspicious sender domain")

    # Check subject
    for word in suspicious_words:
        if word in email["subject"].lower():
            score += 1
            reasons.append("Suspicious subject keyword: " + word)

    # Check attachment
    for ext in suspicious_extensions:
        if email["attachment"].lower().endswith(ext):
            score += 2
            reasons.append("Suspicious attachment: " + ext)

    # Check URL
    if email["url"].startswith("http://"):
        score += 2
        reasons.append("Unsecured or suspicious URL")

    # Store result
    results.append({
        "email": email,
        "score": score,
        "reasons": reasons
    })

# Rank emails by suspicion score
results.sort(key=lambda x: x["score"], reverse=True)

print("EMAIL FORENSIC INVESTIGATION REPORT")
print("=" * 50)

print("\nSHORTLIST OF EMAILS REQUIRING EXAMINATION")

for result in results:
    if result["score"] >= 3:
        email = result["email"]

        print("\nSender:", email["sender"])
        print("Recipient:", email["recipient"])
        print("Timestamp:", email["timestamp"])
        print("Subject:", email["subject"])
        print("Attachment:", email["attachment"])
        print("URL:", email["url"])
        print("Suspicion Score:", result["score"])

        print("Reasons:")
        for reason in result["reasons"]:
            print("-", reason)
```
