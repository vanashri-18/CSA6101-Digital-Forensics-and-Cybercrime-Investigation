**Aim**

To develop a Python-based domain-screening program that evaluates domain names using rule-based indicators such as unusual character combinations, excessive length, numeric substitutions, suspicious keywords, multiple hyphens, and predefined high-risk patterns. The program assigns a score, classifies each domain, and displays the indicators responsible for the classification.

**Algorithm**

Create a list of simulated domain names.

Check each domain for excessive length.

Detect numeric substitutions such as 0, 1, 3, and 4.

Detect suspicious keywords such as login, verify, secure, update, and account.

Count multiple hyphens.

Check predefined high-risk domain patterns.

Check unusual character combinations.

Assign points for every detected indicator.

Calculate the total score.

Classify the domain as LOW, MEDIUM, or HIGH risk.

Display the domain, score, classification, and specific indicators.


```python
# ==============================================
# DOMAIN SCREENING PROGRAM
# ==============================================

import re
import pandas as pd

# ------------------------------------------------
# 1. Simulated Domain Dataset
# ------------------------------------------------

domains = [
    "google.com",
    "amazon.com",
    "secure-login-account.com",
    "paypa1-verify.com",
    "bank---security-login.com",
    "free-prize-winner123.com",
    "microsoft-update-secure-login.com",
    "xn--paypa1-login-9za.com",
    "normalbusiness.org"
]

# ------------------------------------------------
# 2. Screening Rules
# ------------------------------------------------

suspicious_keywords = [
    "login",
    "verify",
    "secure",
    "update",
    "account",
    "password",
    "bank",
    "prize",
    "winner",
    "payment"
]

high_risk_patterns = [
    r"login.*verify",
    r"secure.*account",
    r"free.*prize",
    r"password.*update",
    r"bank.*security"
]

# ------------------------------------------------
# 3. Analyze Domains
# ------------------------------------------------

results = []

for domain in domains:

    score = 0
    indicators = []

    name = domain.lower()

    # --------------------------------------------
    # Excessive length
    # --------------------------------------------

    if len(name) > 25:

        score += 2

        indicators.append(
            "Excessive domain length"
        )

    # --------------------------------------------
    # Numeric substitutions
    # --------------------------------------------

    if re.search(r"[01345]", name):

        score += 1

        indicators.append(
            "Numeric characters/substitution"
        )

    # --------------------------------------------
    # Multiple hyphens
    # --------------------------------------------

    if name.count("-") >= 2:

        score += 2

        indicators.append(
            "Multiple hyphens"
        )

    # --------------------------------------------
    # Suspicious keywords
    # --------------------------------------------

    found_keywords = []

    for keyword in suspicious_keywords:

        if keyword in name:

            found_keywords.append(keyword)

    if found_keywords:

        score += min(
            len(found_keywords), 3
        )

        indicators.append(
            "Suspicious keywords: "
            + ", ".join(found_keywords)
        )

    # --------------------------------------------
    # Unusual character combination
    # --------------------------------------------

    if (
        re.search(r"[a-z]\d", name)
        or re.search(r"\d[a-z]", name)
    ):

        score += 1

        indicators.append(
            "Unusual letter-number combination"
        )

    # --------------------------------------------
    # High-risk predefined patterns
    # --------------------------------------------

    matched_patterns = []

    for pattern in high_risk_patterns:

        if re.search(pattern, name):

            matched_patterns.append(pattern)

    if matched_patterns:

        score += 3

        indicators.append(
            "Predefined high-risk pattern"
        )

    # --------------------------------------------
    # Risk Classification
    # --------------------------------------------

    if score >= 6:

        classification = "HIGH"

    elif score >= 3:

        classification = "MEDIUM"

    else:

        classification = "LOW"

    if not indicators:

        indicators.append(
            "No suspicious indicators detected"
        )

    results.append({
        "Domain": domain,
        "Score": score,
        "Classification": classification,
        "Indicators": "; ".join(indicators)
    })

# ------------------------------------------------
# 4. Create DataFrame
# ------------------------------------------------

report = pd.DataFrame(results)

# ------------------------------------------------
# 5. Display Results
# ------------------------------------------------

print("=" * 110)
print("                         DOMAIN SCREENING REPORT")
print("=" * 110)

print(
    report.to_string(
        index=False
    )
)

# ------------------------------------------------
# 6. Summary
# ------------------------------------------------

print("\n" + "=" * 110)
print("                           SUMMARY")
print("=" * 110)

print(
    "Total Domains :",
    len(report)
)

print(
    "HIGH Risk     :",
    len(
        report[
            report["Classification"] == "HIGH"
        ]
    )
)

print(
    "MEDIUM Risk   :",
    len(
        report[
            report["Classification"] == "MEDIUM"
        ]
    )
)

print(
    "LOW Risk      :",
    len(
        report[
            report["Classification"] == "LOW"
        ]
    )
)

# ------------------------------------------------
# 7. Save Report
# ------------------------------------------------

report.to_csv(
    "/content/domain_screening_report.csv",
    index=False
)

print(
    "\nReport saved as:"
    " /content/domain_screening_report.csv"
)

print("\nDomain screening completed.")
```

**Result**

The Python program successfully screened the simulated domains using multiple rule-based indicators. Domains containing suspicious keywords, multiple hyphens, numeric substitutions, excessive length, unusual character combinations, or predefined high-risk patterns received higher scores. The final report clearly shows the domain, score, risk classification, and exact indicators responsible for each result. A high score is an indicator for further investigation and does not by itself prove that a domain is malicious.
