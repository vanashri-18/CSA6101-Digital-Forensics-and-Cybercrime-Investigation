**Aim**

To develop a Python application that guides an investigator through the five major stages of a digital forensic investigation, records case information and observations at each stage, and generates a structured final summary showing how the collected information contributes to the overall investigation.

Algorithm

1. Identify Case:
Enter case ID, investigator name, incident type, and initial observations.:

2. Preserve Evidence:
Record evidence details, preservation method, and integrity observations.

3. Collect Evidence:
Enter evidence ID, evidence type, source, and collection details.

4. Examine Evidence:
Record tools used, artifacts identified, and examination observations.

5. Analyze Findings:
Document important findings, indicators, timelines, and conclusions.

6. Report Results:
Enter final conclusion and recommended actions.

7. Generate Summary:
Combine information from all five stages into a structured investigation report.

```python
from datetime import datetime

# ---------------------------------------------------------
# DIGITAL FORENSIC INVESTIGATION WORKFLOW
# Five Major Stages:
# 1. Identification
# 2. Preservation
# 3. Collection
# 4. Examination
# 5. Analysis & Reporting
# ---------------------------------------------------------


def get_input(message):
    """Read investigator input."""
    value = input(message).strip()

    if value == "":
        return "Not provided"

    return value


def display_stage(number, title):
    print("\n" + "=" * 70)
    print(f"STAGE {number}: {title}")
    print("=" * 70)


# ---------------------------------------------------------
# START INVESTIGATION
# ---------------------------------------------------------

print("=" * 70)
print("        DIGITAL FORENSIC INVESTIGATION SYSTEM")
print("=" * 70)

case_id = get_input("Enter Case ID: ")
investigator = get_input("Enter Investigator Name: ")

start_time = datetime.now()


# ---------------------------------------------------------
# STAGE 1 - IDENTIFICATION
# ---------------------------------------------------------

display_stage(1, "IDENTIFICATION")

case_type = get_input("Incident Type: ")
initial_observation = get_input(
    "Initial Observation: "
)

suspected_source = get_input(
    "Suspected Evidence Source: "
)

identification = {
    "Incident Type": case_type,
    "Initial Observation": initial_observation,
    "Suspected Source": suspected_source
}


# ---------------------------------------------------------
# STAGE 2 - PRESERVATION
# ---------------------------------------------------------

display_stage(2, "PRESERVATION")

evidence_condition = get_input(
    "Evidence Condition: "
)

preservation_method = get_input(
    "Preservation Method: "
)

integrity_action = get_input(
    "Integrity Action (e.g., SHA-256 hashing): "
)

preservation = {
    "Evidence Condition": evidence_condition,
    "Preservation Method": preservation_method,
    "Integrity Action": integrity_action
}


# ---------------------------------------------------------
# STAGE 3 - COLLECTION
# ---------------------------------------------------------

display_stage(3, "COLLECTION")

evidence_id = get_input("Evidence ID: ")
evidence_type = get_input("Evidence Type: ")
evidence_source = get_input("Evidence Source: ")
collection_method = get_input("Collection Method: ")

collection = {
    "Evidence ID": evidence_id,
    "Evidence Type": evidence_type,
    "Evidence Source": evidence_source,
    "Collection Method": collection_method
}


# ---------------------------------------------------------
# STAGE 4 - EXAMINATION
# ---------------------------------------------------------

display_stage(4, "EXAMINATION")

tools_used = get_input(
    "Forensic Tools Used: "
)

artifacts_found = get_input(
    "Artifacts Identified: "
)

examination_observation = get_input(
    "Examination Observation: "
)

examination = {
    "Tools Used": tools_used,
    "Artifacts Identified": artifacts_found,
    "Observation": examination_observation
}


# ---------------------------------------------------------
# STAGE 5 - ANALYSIS
# ---------------------------------------------------------

display_stage(5, "ANALYSIS")

important_finding = get_input(
    "Important Finding: "
)

ioc = get_input(
    "Indicators of Compromise / Relevant Evidence: "
)

timeline_event = get_input(
    "Important Timeline Event: "
)

analysis_conclusion = get_input(
    "Analytical Conclusion: "
)

analysis = {
    "Important Finding": important_finding,
    "Indicators": ioc,
    "Timeline Event": timeline_event,
    "Conclusion": analysis_conclusion
}


# ---------------------------------------------------------
# REPORTING
# ---------------------------------------------------------

display_stage(6, "REPORTING")

final_conclusion = get_input(
    "Final Investigation Conclusion: "
)

recommendation = get_input(
    "Recommended Action: "
)

report_time = datetime.now()

reporting = {
    "Final Conclusion": final_conclusion,
    "Recommendation": recommendation
}


# ---------------------------------------------------------
# FINAL STRUCTURED SUMMARY
# ---------------------------------------------------------

print("\n\n")
print("=" * 70)
print("             DIGITAL FORENSIC CASE REPORT")
print("=" * 70)

print("\nCASE INFORMATION")
print("-" * 70)

print("Case ID       :", case_id)
print("Investigator  :", investigator)
print(
    "Investigation Started:",
    start_time.strftime("%Y-%m-%d %H:%M:%S")
)
print(
    "Report Generated:",
    report_time.strftime("%Y-%m-%d %H:%M:%S")
)


# ---------------------------------------------------------
# IDENTIFICATION SUMMARY
# ---------------------------------------------------------

print("\n" + "=" * 70)
print("1. IDENTIFICATION")
print("=" * 70)

for key, value in identification.items():
    print(f"{key}: {value}")


# ---------------------------------------------------------
# PRESERVATION SUMMARY
# ---------------------------------------------------------

print("\n" + "=" * 70)
print("2. PRESERVATION")
print("=" * 70)

for key, value in preservation.items():
    print(f"{key}: {value}")


# ---------------------------------------------------------
# COLLECTION SUMMARY
# ---------------------------------------------------------

print("\n" + "=" * 70)
print("3. COLLECTION")
print("=" * 70)

for key, value in collection.items():
    print(f"{key}: {value}")


# ---------------------------------------------------------
# EXAMINATION SUMMARY
# ---------------------------------------------------------

print("\n" + "=" * 70)
print("4. EXAMINATION")
print("=" * 70)

for key, value in examination.items():
    print(f"{key}: {value}")


# ---------------------------------------------------------
# ANALYSIS SUMMARY
# ---------------------------------------------------------

print("\n" + "=" * 70)
print("5. ANALYSIS")
print("=" * 70)

for key, value in analysis.items():
    print(f"{key}: {value}")


# ---------------------------------------------------------
# REPORTING SUMMARY
# ---------------------------------------------------------

print("\n" + "=" * 70)
print("6. REPORTING")
print("=" * 70)

for key, value in reporting.items():
    print(f"{key}: {value}")


# ---------------------------------------------------------
# INVESTIGATION FLOW
# ---------------------------------------------------------

print("\n" + "=" * 70)
print("INVESTIGATION FLOW")
print("=" * 70)

print("Identification")
print("      ↓")
print("Evidence identified from:", suspected_source)

print("      ↓")
print("Preservation")
print("Integrity protection:", integrity_action)

print("      ↓")
print("Collection")
print("Evidence collected:", evidence_id)

print("      ↓")
print("Examination")
print("Artifacts identified:", artifacts_found)

print("      ↓")
print("Analysis")
print("Important finding:", important_finding)

print("      ↓")
print("Reporting")
print("Final conclusion:", final_conclusion)

print("\n" + "=" * 70)
print("FORENSIC INVESTIGATION COMPLETED")
print("=" * 70)
```

**Result**

The Python application successfully guided the investigator through the five major forensic stages: Identification, Preservation, Collection, Examination, and Analysis, followed by Reporting. It collected case information and investigative observations at each stage and generated a structured final summary. The investigation flow demonstrates how identified evidence is preserved, collected, examined, analyzed, and ultimately converted into forensic findings and a final report.
