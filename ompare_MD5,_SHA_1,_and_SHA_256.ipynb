**Aim**

To write a Python program that calculates and compares the MD5, SHA-1, and SHA-256 hash values of a digital evidence file to verify its integrity.

**Algorithm**

Import the hashlib module.

Upload the digital evidence file.

Read the file in binary mode.

Create MD5, SHA-1, and SHA-256 hash objects.

Read the file in chunks and update each hash object.

Generate the hexadecimal hash values.

Display all three hash values.

Compare the hash lengths and discuss their security.

```python
# Experiment 14
# Calculate and Compare MD5, SHA-1, and SHA-256 Hash Values

import hashlib
from google.colab import files

# Upload evidence file
uploaded = files.upload()

# Get uploaded filename
filename = list(uploaded.keys())[0]

# Function to calculate hashes
def calculate_hashes(file_path):
    md5 = hashlib.md5()
    sha1 = hashlib.sha1()
    sha256 = hashlib.sha256()

    with open(file_path, "rb") as file:
        while True:
            chunk = file.read(4096)
            if not chunk:
                break
            md5.update(chunk)
            sha1.update(chunk)
            sha256.update(chunk)

    return md5.hexdigest(), sha1.hexdigest(), sha256.hexdigest()

# Calculate hash values
md5_hash, sha1_hash, sha256_hash = calculate_hashes(filename)

# Display results
print("Digital Evidence File:", filename)
print("-" * 60)
print("MD5 Hash     :", md5_hash)
print("SHA-1 Hash   :", sha1_hash)
print("SHA-256 Hash :", sha256_hash)

# Compare hash lengths
print("\nHash Comparison")
print("-" * 60)
print("MD5 Length     :", len(md5_hash), "characters")
print("SHA-1 Length   :", len(sha1_hash), "characters")
print("SHA-256 Length :", len(sha256_hash), "characters")

print("\nSecurity Comparison:")
print("MD5     : Fast but vulnerable to collisions.")
print("SHA-1   : More secure than MD5 but deprecated due to collision attacks.")
print("SHA-256 : Highly secure and recommended for digital forensic investigations.")
```

**Result**

The Python program successfully calculated the MD5, SHA-1, and SHA-256 hash values of the uploaded digital evidence file. The generated hashes can be used to verify file integrity. Among the three algorithms, SHA-256 provides the highest level of security and is the preferred choice for digital forensic investigations.
