**Aim**

To develop a Python program that accepts investigation findings from an investigator and automatically organizes them into a structured digital forensic report containing case information, evidence examined, examination procedures, observations, findings, conclusion, and report-generation timestamp.

**Algorithm**

Start the Python program.

Collect case information from the investigator.

Collect details of the evidence examined.

Record the examination procedures performed.

Record observations made during the investigation.

Record the investigation findings.

Record the final conclusion.

Generate the current date and time automatically.

Organize all information into clearly separated report sections.

Display and save the completed digital forensic report.


```python
# ==============================================
# DIGITAL FORENSIC REPORT GENERATOR
# ==============================================

from datetime import datetime

print("=" * 70)
print("           DIGITAL FORENSIC REPORT GENERATOR")
print("=" * 70)

# ------------------------------------------------
# 1. Collect Case Information
# ------------------------------------------------

case_id = input("\nEnter Case ID: ")
case_title = input("Enter Case Title: ")
investigator = input("Enter Investigator Name: ")
organization = input("Enter Organization: ")

# ------------------------------------------------
# 2. Evidence Information
# ------------------------------------------------

print("\n--- Evidence Examined ---")

evidence = input(
    "Enter evidence examined: "
)

evidence_id = input(
    "Enter Evidence ID / Reference: "
)

# ------------------------------------------------
# 3. Examination Procedure
# ------------------------------------------------

print("\n--- Examination Procedure ---")

procedure = input(
    "Enter examination procedure performed: "
)

# ------------------------------------------------
# 4. Observations
# ------------------------------------------------

print("\n--- Observations ---")

observations = input(
    "Enter investigation observations: "
)

# ------------------------------------------------
# 5. Findings
# ------------------------------------------------

print("\n--- Findings ---")

findings = input(
    "Enter investigation findings: "
)

# ------------------------------------------------
# 6. Conclusion
# ------------------------------------------------

print("\n--- Conclusion ---")

conclusion = input(
    "Enter final conclusion: "
)

# ------------------------------------------------
# 7. Generate Timestamp
# ------------------------------------------------

report_time = datetime.now().strftime(
    "%Y-%m-%d %H:%M:%S"
)

# ------------------------------------------------
# 8. Generate Report
# ------------------------------------------------

report = f"""
======================================================================
                    DIGITAL FORENSIC REPORT
======================================================================

CASE INFORMATION
----------------------------------------------------------------------
Case ID              : {case_id}
Case Title           : {case_title}
Investigator         : {investigator}
Organization         : {organization}

EVIDENCE EXAMINED
----------------------------------------------------------------------
Evidence ID          : {evidence_id}
Evidence Description : {evidence}

EXAMINATION PROCEDURES
----------------------------------------------------------------------
{procedure}

OBSERVATIONS
----------------------------------------------------------------------
{observations}

FINDINGS
----------------------------------------------------------------------
{findings}

CONCLUSION
----------------------------------------------------------------------
{conclusion}

REPORT INFORMATION
----------------------------------------------------------------------
Report Generated     : {report_time}

======================================================================
                    END OF FORENSIC REPORT
======================================================================
"""

# ------------------------------------------------
# 9. Display Report
# ------------------------------------------------

print("\n")
print(report)

# ------------------------------------------------
# 10. Save Report
# ------------------------------------------------

filename = f"{case_id}_forensic_report.txt"

with open(filename, "w") as file:
    file.write(report)

print(f"\nReport saved successfully as: {filename}")
```

**Result**

The Python program successfully collected investigation details from the investigator and automatically generated a structured digital forensic report. The report clearly separated case information, evidence examined, examination procedures, observations, findings, and conclusion, and automatically included the date and time of report generation. The completed report was also saved as a text file, providing a reusable record of the forensic investigation.
