```python
# File Type Verification Utility

files = [
    {
        "name": "report.pdf",
        "signature": "%PDF",
        "content": "Document content"
    },
    {
        "name": "photo.jpg",
        "signature": "FFD8FF",
        "content": "Image data"
    },
    {
        "name": "backup.zip",
        "signature": "504B0304",
        "content": "Archive data"
    },
    {
        "name": "program.exe",
        "signature": "4D5A",
        "content": "Executable code"
    },
    {
        "name": "holiday.jpg",
        "signature": "4D5A",
        "content": "Executable code"
    }
]

# Expected information for file types
file_types = {
    ".pdf": {
        "signature": "%PDF",
        "content": "Document content",
        "type": "PDF Document"
    },
    ".jpg": {
        "signature": "FFD8FF",
        "content": "Image data",
        "type": "JPEG Image"
    },
    ".zip": {
        "signature": "504B0304",
        "content": "Archive data",
        "type": "ZIP Archive"
    },
    ".exe": {
        "signature": "4D5A",
        "content": "Executable code",
        "type": "Executable File"
    }
}

print("FILE TYPE VERIFICATION REPORT")
print("=" * 50)

for file in files:

    # Get extension
    extension = "." + file["name"].split(".")[-1]

    print("\nFilename:", file["name"])
    print("Extension:", extension)
    print("File Signature:", file["signature"])
    print("Content Type:", file["content"])

    if extension in file_types:
        expected = file_types[extension]

        # Check consistency
        if (file["signature"] == expected["signature"] and
                file["content"] == expected["content"]):

            print("Classification:", expected["type"])
            print("Status: CONSISTENT")
            print("Warning: No file type inconsistency detected.")

        else:
            print("Classification: SUSPICIOUS")
            print("Status: INCONSISTENT")
            print("Warning: Filename extension does not match")
            print("the file signature or content characteristics!")

    else:
        print("Classification: UNKNOWN")
        print("Warning: Unsupported file extension.")
```
