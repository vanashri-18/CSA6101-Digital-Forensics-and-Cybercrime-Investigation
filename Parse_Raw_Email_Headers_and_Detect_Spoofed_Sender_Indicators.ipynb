**Aim**

To parse raw email headers and identify indicators of a spoofed sender by examining fields such as From, Return-Path, Received, and Authentication-Results using Python.

**Algorithm**

Import the required Python libraries.

Upload the raw email header file.

Read the email header line by line.

Extract important header fields such as From, Return-Path, Received, and Authentication-Results.

Compare the From address with the Return-Path address.

Check whether SPF, DKIM, or DMARC authentication has failed.

Display possible spoofing indicators if mismatches or authentication failures are detected.

```python
from google.colab import files

print("Upload the raw email header file:")
uploaded = files.upload()

filename = list(uploaded.keys())[0]

headers = {}

# Read file safely
with open(filename, "r", encoding="latin-1") as file:
    for line in file:
        if ":" in line:
            key, value = line.split(":", 1)
            headers[key.strip()] = value.strip()

print("\nEmail Header Analysis")
print("----------------------------")

for field in ["From", "Return-Path", "Received", "Authentication-Results"]:
    print(f"{field}: {headers.get(field, 'Not Found')}")

print("\nSpoofing Indicators:")

spoof = False

if headers.get("From") != headers.get("Return-Path"):
    print("- From address and Return-Path do not match.")
    spoof = True

auth = headers.get("Authentication-Results", "").lower()

if "spf=fail" in auth:
    print("- SPF authentication failed.")
    spoof = True

if "dkim=fail" in auth:
    print("- DKIM authentication failed.")
    spoof = True

if "dmarc=fail" in auth:
    print("- DMARC authentication failed.")
    spoof = True

if not spoof:
    print("No obvious spoofing indicators detected.")
```

**Result**

The Python program successfully parsed the raw email headers and identified spoofed sender indicators by detecting mismatched email addresses and failed SPF, DKIM, and DMARC authentication checks. This helps in identifying potentially fraudulent or phishing emails.
