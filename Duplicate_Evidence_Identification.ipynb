**Aim**

To develop a Python program that calculates SHA-256 hashes for all files in a selected directory and groups files having identical hash values, thereby distinguishing unique evidence files from duplicate copies.



**Algorithm**

Select the directory containing evidence files.

Identify all files in the directory.
Read each file in binary mode.

Calculate the SHA-256 hash for every file.

Group files having the same SHA-256 hash.

If a hash occurs only once, classify the file as Unique Evidence.

If a hash occurs more than once, classify the files as Duplicate Evidence.

Display the SHA-256 value for each group.

Display all files belonging to each duplicate group.

Generate a final summary of unique and duplicate evidence.

```python

# ==============================================
# DUPLICATE DIGITAL EVIDENCE DETECTOR
# ==============================================

import os
import hashlib
from collections import defaultdict

# ------------------------------------------------
# 1. Select Evidence Directory
# ------------------------------------------------

directory = input(
    "Enter evidence directory path: "
).strip()

if not os.path.isdir(directory):

    print("\nERROR: Directory does not exist.")

else:

    # ------------------------------------------------
    # 2. SHA-256 Function
    # ------------------------------------------------

    def calculate_sha256(filepath):

        sha256 = hashlib.sha256()

        with open(filepath, "rb") as file:

            while True:

                data = file.read(4096)

                if not data:
                    break

                sha256.update(data)

        return sha256.hexdigest()

    # ------------------------------------------------
    # 3. Calculate Hashes
    # ------------------------------------------------

    hash_groups = defaultdict(list)

    for root, folders, files in os.walk(directory):

        for filename in files:

            filepath = os.path.join(root, filename)

            try:

                file_hash = calculate_sha256(filepath)

                hash_groups[file_hash].append(
                    filepath
                )

            except Exception as error:

                print(
                    f"Could not process {filepath}: {error}"
                )

    # ------------------------------------------------
    # 4. Display Results
    # ------------------------------------------------

    print("\n" + "=" * 80)
    print("              DIGITAL EVIDENCE HASH ANALYSIS")
    print("=" * 80)

    unique_count = 0
    duplicate_count = 0
    duplicate_groups = 0

    for file_hash, files in hash_groups.items():

        if len(files) == 1:

            unique_count += 1

        else:

            duplicate_groups += 1
            duplicate_count += len(files)

            print("\n" + "-" * 80)
            print("DUPLICATE EVIDENCE GROUP")
            print("-" * 80)

            print("SHA-256:")
            print(file_hash)

            print("\nFiles with identical content:")

            for file in files:
                print("  ", file)

    # ------------------------------------------------
    # 5. Unique Evidence
    # ------------------------------------------------

    print("\n" + "=" * 80)
    print("                  UNIQUE EVIDENCE")
    print("=" * 80)

    for file_hash, files in hash_groups.items():

        if len(files) == 1:

            print("\nFile:", files[0])
            print("SHA-256:", file_hash)

    # ------------------------------------------------
    # 6. Summary
    # ------------------------------------------------

    total_files = sum(
        len(files) for files in hash_groups.values()
    )

    print("\n" + "=" * 80)
    print("                    SUMMARY")
    print("=" * 80)

    print("Total Files       :", total_files)
    print("Unique Files      :", unique_count)
    print("Duplicate Groups  :", duplicate_groups)
    print("Duplicate Copies  :", duplicate_count)

    print("\nAnalysis completed successfully.")
```

**Result**

The Python program successfully calculated SHA-256 hashes for the evidence files and grouped files with identical hash values. Files having the same SHA-256 value were identified as duplicate copies, while files with unique hashes were classified as unique evidence. This approach helps forensic investigators identify redundant evidence copies efficiently while preserving the distinction between identical and different digital evidence.
