**Aim**

To develop a Python program that scans filenames for potentially deceptive extension patterns, including multiple extensions, executable extensions following document extensions, excessive extension combinations, and predefined suspicious patterns, and generates a warning report without modifying or deleting any files.

**Algorithm**

Select the directory containing the files.

Read all filenames in the selected directory.

Convert filenames to lowercase for pattern analysis.

Split each filename into its extensions.
Check for multiple extensions.

Check for executable extensions following document extensions.

Check for excessive numbers of extensions.

Check for other predefined suspicious filename patterns.

Record the original filename and the reason for each warning.

Generate and display the suspicious filename report.

```python
# ==============================================
# SUSPICIOUS FILENAME / EXTENSION ANALYZER
# ==============================================

import os
import re
import pandas as pd

# ------------------------------------------------
# 1. Select Directory
# ------------------------------------------------

directory = input(
    "Enter directory path: "
).strip()

if not os.path.isdir(directory):

    print("\nERROR: Directory does not exist.")

else:

    # ------------------------------------------------
    # 2. Predefined Extension Lists
    # ------------------------------------------------

    document_ext = {
        ".pdf", ".doc", ".docx", ".xls",
        ".xlsx", ".ppt", ".pptx", ".txt",
        ".jpg", ".jpeg", ".png"
    }

    executable_ext = {
        ".exe", ".scr", ".bat", ".cmd",
        ".com", ".msi", ".ps1", ".vbs",
        ".js"
    }

    suspicious_ext = {
        ".zip", ".rar", ".7z", ".iso"
    }

    # ------------------------------------------------
    # 3. Analyze Filenames
    # ------------------------------------------------

    warnings = []

    for filename in os.listdir(directory):

        filepath = os.path.join(directory, filename)

        if not os.path.isfile(filepath):
            continue

        lower_name = filename.lower()

        # Extract extensions
        extensions = re.findall(
            r'\.[a-z0-9]+',
            lower_name
        )

        reasons = []

        # --------------------------------------------
        # Multiple extensions
        # --------------------------------------------

        if len(extensions) >= 2:

            reasons.append(
                "Multiple file extensions detected"
            )

        # --------------------------------------------
        # Document + executable pattern
        # --------------------------------------------

        if len(extensions) >= 2:

            first_ext = extensions[-2]
            last_ext = extensions[-1]

            if (
                first_ext in document_ext
                and last_ext in executable_ext
            ):

                reasons.append(
                    "Executable extension follows a document extension"
                )

        # --------------------------------------------
        # Excessive extensions
        # --------------------------------------------

        if len(extensions) >= 3:

            reasons.append(
                f"Excessive extension combination ({len(extensions)} extensions)"
            )

        # --------------------------------------------
        # Suspicious archive + executable
        # --------------------------------------------

        if len(extensions) >= 2:

            if (
                extensions[-2] in suspicious_ext
                and extensions[-1] in executable_ext
            ):

                reasons.append(
                    "Executable extension follows an archive extension"
                )

        # --------------------------------------------
        # Double-dot pattern
        # --------------------------------------------

        if ".." in lower_name:

            reasons.append(
                "Multiple consecutive dots detected"
            )

        # --------------------------------------------
        # Long extension chain
        # --------------------------------------------

        if lower_name.count(".") >= 4:

            reasons.append(
                "Unusually long filename extension chain"
            )

        # --------------------------------------------
        # Record Warning
        # --------------------------------------------

        if reasons:

            warnings.append({
                "Filename": filename,
                "Reason": "; ".join(reasons)
            })

    # ------------------------------------------------
    # 4. Display Warning Report
    # ------------------------------------------------

    print("\n" + "=" * 90)
    print("              SUSPICIOUS FILENAME ANALYSIS")
    print("=" * 90)

    if warnings:

        report = pd.DataFrame(warnings)

        print("\nPotentially suspicious filenames:\n")

        print(
            report.to_string(index=False)
        )

        # --------------------------------------------
        # Save Report
        # --------------------------------------------

        report.to_csv(
            "/content/suspicious_filename_report.csv",
            index=False
        )

        print("\n" + "-" * 90)
        print("Warning report saved as:")
        print("/content/suspicious_filename_report.csv")

    else:

        print(
            "\nNo predefined suspicious filename patterns detected."
        )

    print("\nAnalysis completed.")
```

**Result**

The Python program successfully scanned filenames and identified potentially deceptive extension patterns such as multiple extensions, executable extensions following document/archive extensions, excessive extension combinations, and consecutive dots. A warning report containing the original filename and the reason for flagging was generated and saved as a CSV file. The program only reports suspicious patterns and does not delete, rename, or modify any files, allowing the investigator to perform further examination safely.
