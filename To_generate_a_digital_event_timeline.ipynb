**Aim**

To write a Python program to generate a digital event timeline by combining file timestamps and system log records, normalizing timestamps, and arranging all events in chronological order.

**Algorithm**

Import the required Python libraries.

Create simulated file timestamp records.

Create simulated system log records.

Convert both datasets into Pandas DataFrames.

Normalize the timestamp formats.

Combine file and system events.

Sort all events according to their timestamps.

Display the reconstructed digital event timeline.

Identify important events from the timeline.

```python
import pandas as pd
from io import StringIO

# =====================================================
# 1. Simulated File Timestamp Records
# =====================================================

file_data = """Timestamp,EventType,AffectedObject,Description
2026-08-18 09:10:15,Created,report.docx,New report file created
18/08/2026 09:25:30,Modified,report.docx,Report file modified
Aug 18 2026 09:40:45,Accessed,customer.xlsx,Customer data accessed
2026/08/18 10:05:20,Deleted,backup.zip,Backup file deleted
2026-08-18 10:20:35,Created,suspicious.exe,Suspicious executable created
"""

# =====================================================
# 2. Simulated System Log Records
# =====================================================

system_data = """Timestamp,EventType,AffectedObject,Description
18-Aug-2026 09:15:10,Login,User1,User successfully logged in
2026/08/18 09:30:25,Process Start,powershell.exe,PowerShell process started
2026-08-18T09:50:40,USB Connected,USB001,USB device connected
18/08/2026 10:10:55,Process Start,suspicious.exe,Suspicious process executed
2026-08-18 10:30:10,Network Connection,192.168.1.50,External network connection detected
2026-08-18 10:45:20,Logout,User1,User logged out
"""

# =====================================================
# 3. Read the Data
# =====================================================

file_df = pd.read_csv(StringIO(file_data))
system_df = pd.read_csv(StringIO(system_data))

# Add source information
file_df["Source"] = "File Timestamp"
system_df["Source"] = "System Log"

# =====================================================
# 4. Normalize Timestamp Formats
# =====================================================

file_df["Timestamp"] = pd.to_datetime(
    file_df["Timestamp"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

system_df["Timestamp"] = pd.to_datetime(
    system_df["Timestamp"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

# =====================================================
# 5. Combine File and System Events
# =====================================================

timeline = pd.concat(
    [file_df, system_df],
    ignore_index=True
)

# =====================================================
# 6. Sort Events Chronologically
# =====================================================

timeline = timeline.sort_values(
    by="Timestamp"
).reset_index(drop=True)

# =====================================================
# 7. Add Event Number
# =====================================================

timeline.insert(
    0,
    "EventNo",
    range(1, len(timeline) + 1)
)

# =====================================================
# 8. Display Digital Event Timeline
# =====================================================

print("=" * 80)
print("             DIGITAL EVENT TIMELINE")
print("=" * 80)

print(
    timeline[
        [
            "EventNo",
            "Timestamp",
            "Source",
            "EventType",
            "AffectedObject",
            "Description"
        ]
    ].to_string(index=False)
)

# =====================================================
# 9. Identify Important Events
# =====================================================

important_keywords = [
    "deleted",
    "suspicious",
    "usb",
    "external",
    "powershell"
]

important_events = timeline[
    timeline["Description"].str.lower().apply(
        lambda description: any(
            keyword in description
            for keyword in important_keywords
        )
    )
]

print("\n" + "=" * 80)
print("                 IMPORTANT EVENTS")
print("=" * 80)

if important_events.empty:
    print("No important events detected.")

else:
    print(
        important_events[
            [
                "EventNo",
                "Timestamp",
                "Source",
                "EventType",
                "AffectedObject",
                "Description"
            ]
        ].to_string(index=False)
    )

# =====================================================
# 10. Summary
# =====================================================

print("\n" + "=" * 80)
print("                    SUMMARY")
print("=" * 80)

print("Total Events       :", len(timeline))
print(
    "File Events        :",
    len(file_df)
)
print(
    "System Log Events  :",
    len(system_df)
)
print(
    "Important Events   :",
    len(important_events)
)

print("\nDigital event timeline generated successfully.")
```

**Result**

The Python program successfully generated a digital event timeline by combining file timestamps and system log records. Different timestamp formats were normalized, and all events were arranged chronologically. The resulting timeline helps forensic investigators understand the sequence of system activities and identify potentially suspicious events.
