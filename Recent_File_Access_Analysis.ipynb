**Aim**

To develop a Python program that analyzes a simulated operating-system activity log, identifies the most recently accessed files, sorts them in reverse chronological order, and allows investigators to filter the results by a specified time range and username.

**Algorithm**

Create or load the simulated OS activity log.

Read the username, file path, and access timestamp.

Convert the timestamp into a datetime format.

Accept the investigator's start and end time.

Optionally accept a username filter.

Select records within the specified time range.

Apply the username filter when provided.

Sort the selected files from newest to oldest.

Display the recently accessed files.

Generate an activity summary showing the number of accesses and users involved.

```python
# ==============================================
# RECENT FILE ACCESS ANALYZER
# ==============================================

import pandas as pd

# ------------------------------------------------
# 1. Simulated Operating-System Activity Log
# ------------------------------------------------

data = [
    ["2026-08-25 08:10:00", "Alice", "C:/Documents/report.docx"],
    ["2026-08-25 08:25:00", "Bob", "C:/Downloads/setup.exe"],
    ["2026-08-25 08:40:00", "Alice", "C:/Documents/evidence.pdf"],
    ["2026-08-25 09:00:00", "Charlie", "C:/Desktop/notes.txt"],
    ["2026-08-25 09:15:00", "Alice", "C:/Documents/report.docx"],
    ["2026-08-25 09:30:00", "Bob", "C:/Downloads/data.zip"],
    ["2026-08-25 09:45:00", "Alice", "C:/Documents/log.txt"],
    ["2026-08-25 10:00:00", "Charlie", "C:/Desktop/image.jpg"],
    ["2026-08-25 10:20:00", "Bob", "C:/Downloads/setup.exe"],
    ["2026-08-25 10:40:00", "Alice", "C:/Documents/evidence.pdf"]
]

df = pd.DataFrame(
    data,
    columns=["Timestamp", "Username", "File_Path"]
)

# Convert timestamps
df["Timestamp"] = pd.to_datetime(
    df["Timestamp"]
)

print("=" * 85)
print("                 RECENT FILE ACCESS ANALYZER")
print("=" * 85)

# ------------------------------------------------
# 2. Investigator Input
# ------------------------------------------------

start_time = input(
    "\nEnter start time (YYYY-MM-DD HH:MM:SS): "
)

end_time = input(
    "Enter end time (YYYY-MM-DD HH:MM:SS): "
)

username = input(
    "Enter username to filter (press Enter for all users): "
).strip()

start_time = pd.to_datetime(start_time)
end_time = pd.to_datetime(end_time)

# ------------------------------------------------
# 3. Filter by Time Range
# ------------------------------------------------

filtered = df[
    (df["Timestamp"] >= start_time) &
    (df["Timestamp"] <= end_time)
].copy()

# ------------------------------------------------
# 4. Optional Username Filter
# ------------------------------------------------

if username:

    filtered = filtered[
        filtered["Username"].str.lower()
        == username.lower()
    ]

# ------------------------------------------------
# 5. Sort Reverse Chronologically
# ------------------------------------------------

filtered = filtered.sort_values(
    by="Timestamp",
    ascending=False
)

# ------------------------------------------------
# 6. Display Results
# ------------------------------------------------

print("\n" + "=" * 85)
print("                 RECENTLY ACCESSED FILES")
print("=" * 85)

if filtered.empty:

    print("\nNo file-access activity found.")

else:

    print(
        filtered.to_string(index=False)
    )

    # ------------------------------------------------
    # 7. Activity Summary
    # ------------------------------------------------

    print("\n" + "-" * 85)
    print("                    ACTIVITY SUMMARY")
    print("-" * 85)

    print(
        "Time Range :",
        start_time,
        "to",
        end_time
    )

    if username:
        print("Username   :", username)
    else:
        print("Username   : All users")

    print(
        "Total Access Events :",
        len(filtered)
    )

    print(
        "Unique Files        :",
        filtered["File_Path"].nunique()
    )

    print(
        "Users Involved      :",
        filtered["Username"].nunique()
    )

    print("\nAccess Count by User:")
    print(
        filtered["Username"]
        .value_counts()
        .to_string()
    )

    print("\nMost Frequently Accessed Files:")

    print(
        filtered["File_Path"]
        .value_counts()
        .to_string()
    )

print("\n" + "=" * 85)
print("File access analysis completed.")
print("=" * 85)
```

**Result**

The Python program successfully analyzed the simulated operating-system activity log and identified files accessed within the investigator-specified time range. The results were arranged in reverse chronological order, with an optional username filter. The activity summary reports the total access events, unique files, users involved, and frequently accessed files, helping investigators identify recent file activity efficiently.
