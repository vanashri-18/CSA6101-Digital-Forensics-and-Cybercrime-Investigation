```python
from datetime import datetime

# Incident response events
events = [
    {
        "time": "2026-08-29 09:00:00",
        "phase": "Detection",
        "event": "Suspicious login activity detected"
    },
    {
        "time": "2026-08-29 09:10:00",
        "phase": "Initial Analysis",
        "event": "Security team analyzed authentication logs"
    },
    {
        "time": "2026-08-29 09:25:00",
        "phase": "Containment",
        "event": "Affected user account was temporarily blocked"
    },
    {
        "time": "2026-08-29 10:00:00",
        "phase": "Eradication",
        "event": "Malicious files were removed from the system"
    },
    {
        "time": "2026-08-29 11:00:00",
        "phase": "Recovery",
        "event": "System services were restored and monitored"
    },
    {
        "time": "2026-08-29 12:00:00",
        "phase": "Follow-Up",
        "event": "Incident report and recommendations were prepared"
    }
]

# Convert timestamps into datetime objects
for item in events:
    item["datetime"] = datetime.strptime(
        item["time"], "%Y-%m-%d %H:%M:%S"
    )

# Sort events chronologically
events.sort(key=lambda x: x["datetime"])

# Display timeline
print("INCIDENT RESPONSE TIMELINE")
print("-" * 50)

for item in events:
    print("Time  :", item["time"])
    print("Phase :", item["phase"])
    print("Event :", item["event"])
    print("-" * 50)
```
