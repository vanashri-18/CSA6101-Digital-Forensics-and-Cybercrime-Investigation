**Aim**

To develop a Python program that performs DNS-based reconnaissance on an investigator-supplied domain, retrieves its resolved IP addresses and standard DNS information, and handles invalid or non-existent domains gracefully.

**Algorithm**

1. Enter Domain:
Accept a domain name from the investigator.

2. Validate Input:
Check whether the entered domain is non-empty and has a valid hostname format.

3. Resolve Domain:
Use Python's standard socket library to perform DNS resolution.

4. Retrieve Addresses:
Obtain the IPv4 and IPv6 addresses associated with the domain.

5. Retrieve Hostname:
Display the canonical/resolved hostname obtained through standard DNS resolution.

6. Handle Errors:
Detect invalid or non-existent domains and display an appropriate message instead of terminating the program.

7. Record Time:
Record the date and time at which the investigation was performed.

8. Display Results:
Present the queried domain, DNS results, investigation time, and status in an investigator-friendly format.

```python
import socket
import re
from datetime import datetime


# ---------------------------------------------------------
# DNS RECONNAISSANCE PROGRAM
# ---------------------------------------------------------

def validate_domain(domain):
    """
    Basic domain-name validation.
    """

    pattern = r"^(?=.{1,253}$)([a-zA-Z0-9-]+\.)+[a-zA-Z]{2,}$"

    return re.match(pattern, domain) is not None


def dns_reconnaissance(domain):

    investigation_time = datetime.now()

    print("\n" + "=" * 65)
    print("             DNS RECONNAISSANCE REPORT")
    print("=" * 65)

    print("Domain Queried :", domain)
    print(
        "Investigation Time:",
        investigation_time.strftime("%Y-%m-%d %H:%M:%S")
    )

    # -----------------------------------------------------
    # Validate domain
    # -----------------------------------------------------

    if not validate_domain(domain):

        print("\nQuery Result   : INVALID DOMAIN FORMAT")
        print("Status         : Investigation stopped safely.")
        print("=" * 65)

        return

    try:

        # -------------------------------------------------
        # Get hostname and IP addresses
        # -------------------------------------------------

        host_name, aliases, addresses = socket.gethostbyname_ex(domain)

        print("\nQuery Result   : DNS RESOLUTION SUCCESSFUL")

        print("\nCanonical Hostname:")
        print(host_name)

        # -------------------------------------------------
        # Display aliases
        # -------------------------------------------------

        print("\nAliases:")

        if aliases:

            for alias in aliases:
                print(" -", alias)

        else:
            print(" - None reported")

        # -------------------------------------------------
        # Display IPv4 addresses
        # -------------------------------------------------

        print("\nIPv4 Addresses:")

        if addresses:

            for ip in addresses:
                print(" -", ip)

        else:
            print(" - No IPv4 address returned")

        # -------------------------------------------------
        # IPv6 resolution
        # -------------------------------------------------

        print("\nIPv6 Addresses:")

        ipv6_results = socket.getaddrinfo(
            domain,
            None,
            socket.AF_INET6
        )

        ipv6_addresses = set()

        for result in ipv6_results:

            ipv6_address = result[4][0]
            ipv6_addresses.add(ipv6_address)

        if ipv6_addresses:

            for ip in sorted(ipv6_addresses):
                print(" -", ip)

        else:
            print(" - No IPv6 address returned")

        print("\nStatus         : RESOLUTION COMPLETED")

    except socket.gaierror:

        print("\nQuery Result   : DOMAIN COULD NOT BE RESOLVED")
        print("Status         : Domain may not exist or DNS resolution failed.")

    except socket.timeout:

        print("\nQuery Result   : DNS QUERY TIMEOUT")
        print("Status         : DNS server did not respond in time.")

    except Exception as error:

        print("\nQuery Result   : UNEXPECTED ERROR")
        print("Status         :", error)

    print("=" * 65)


# ---------------------------------------------------------
# MAIN PROGRAM
# ---------------------------------------------------------

print("DNS-Based Digital Forensic Reconnaissance")

domain = input("\nEnter domain name: ").strip()

dns_reconnaissance(domain)
```

**Result**

The Python program successfully performed DNS-based reconnaissance by accepting an investigator-supplied domain and retrieving its resolved hostname, IPv4 addresses, IPv6 addresses, and aliases using standard DNS resolution. It also recorded the domain queried and investigation date/time and handled invalid or non-existent domains without crashing.
