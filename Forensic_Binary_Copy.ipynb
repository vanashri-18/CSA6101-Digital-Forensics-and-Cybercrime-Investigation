**Aim**

To create an exact binary copy of a selected evidence file, calculate SHA-256 hashes for the original and copied files, and verify the copy using file size and hash comparison.

**Algorithm**

1. Select Evidence:
Enter the path of the evidence file.

2. Read Binary:
Open the original file in binary read mode.

3. Copy Blocks:
Read the file in binary blocks and write the same data to a forensic-copy file.

4. Calculate Hash:
Generate the SHA-256 hash of both files.

5. Compare Size:
Compare the sizes of the original and copied files.

6. Compare Hash:
256 values of both files.

7. Verify Copy:
If both size and hash match, confirm that the forensic copy is an exact representation.

```python
import os
import hashlib
from datetime import datetime
from google.colab import files

# --------------------------------------------------
# DIGITAL FORENSIC BINARY COPY
# --------------------------------------------------

print("=" * 65)
print("        DIGITAL FORENSIC BINARY COPY")
print("=" * 65)

# --------------------------------------------------
# UPLOAD EVIDENCE FILE
# --------------------------------------------------

print("\nSelect the evidence file to upload:")

uploaded = files.upload()

if not uploaded:
    print("\nERROR: No file was selected.")
else:

    # Get uploaded filename
    source_file = list(uploaded.keys())[0]

    print("\nEvidence file selected:", source_file)

    # Create forensic copy filename
    copy_file = source_file + "_FORENSIC_COPY"

    copy_time = datetime.now()

    # --------------------------------------------------
    # CREATE EXACT BINARY COPY
    # --------------------------------------------------

    print("\nCreating forensic copy...")

    with open(source_file, "rb") as source:
        with open(copy_file, "wb") as destination:

            while True:

                data = source.read(4096)

                if not data:
                    break

                destination.write(data)

    print("Forensic copy created successfully.")

    # --------------------------------------------------
    # SHA-256 FUNCTION
    # --------------------------------------------------

    def calculate_sha256(filename):

        sha256 = hashlib.sha256()

        with open(filename, "rb") as file:

            while True:

                data = file.read(4096)

                if not data:
                    break

                sha256.update(data)

        return sha256.hexdigest()

    # --------------------------------------------------
    # FILE SIZE
    # --------------------------------------------------

    original_size = os.path.getsize(source_file)
    copy_size = os.path.getsize(copy_file)

    # --------------------------------------------------
    # CALCULATE HASHES
    # --------------------------------------------------

    original_hash = calculate_sha256(source_file)
    copy_hash = calculate_sha256(copy_file)

    # --------------------------------------------------
    # DISPLAY DETAILS
    # --------------------------------------------------

    print("\n" + "=" * 65)
    print("FORENSIC COPY DETAILS")
    print("=" * 65)

    print("Original File :", source_file)
    print("Copy File     :", copy_file)

    print(
        "Copy Time     :",
        copy_time.strftime("%Y-%m-%d %H:%M:%S")
    )

    print("\nOriginal Size :", original_size, "bytes")
    print("Copy Size     :", copy_size, "bytes")

    print("\nOriginal SHA-256:")
    print(original_hash)

    print("\nCopy SHA-256:")
    print(copy_hash)

    # --------------------------------------------------
    # SIZE COMPARISON
    # --------------------------------------------------

    print("\n" + "=" * 65)
    print("SIZE COMPARISON")
    print("=" * 65)

    if original_size == copy_size:

        print("Original Size :", original_size, "bytes")
        print("Copy Size     :", copy_size, "bytes")
        print("Result        : MATCH")

        size_match = True

    else:

        print("Original Size :", original_size, "bytes")
        print("Copy Size     :", copy_size, "bytes")
        print("Result        : MISMATCH")

        size_match = False

    # --------------------------------------------------
    # HASH COMPARISON
    # --------------------------------------------------

    print("\n" + "=" * 65)
    print("HASH COMPARISON")
    print("=" * 65)

    if original_hash == copy_hash:

        print("Original Hash :")
        print(original_hash)

        print("\nCopy Hash :")
        print(copy_hash)

        print("\nResult: SHA-256 HASH MATCH")

        hash_match = True

    else:

        print("Original Hash :")
        print(original_hash)

        print("\nCopy Hash :")
        print(copy_hash)

        print("\nResult: SHA-256 HASH MISMATCH")

        hash_match = False

    # --------------------------------------------------
    # FINAL VERIFICATION
    # --------------------------------------------------

    print("\n" + "=" * 65)
    print("FINAL VERIFICATION")
    print("=" * 65)

    if size_match and hash_match:

        print("STATUS: VERIFIED")
        print()
        print("The forensic copy is an exact")
        print("representation of the original file.")
        print()
        print("File Size Match : YES")
        print("Hash Match      : YES")

    else:

        print("STATUS: VERIFICATION FAILED")
        print()
        print("The forensic copy does not exactly")
        print("match the original evidence file.")
        print()
        print("File Size Match :",
              "YES" if size_match else "NO")

        print("Hash Match      :",
              "YES" if hash_match else "NO")

    print("\n" + "=" * 65)
    print("FORENSIC VERIFICATION COMPLETED")
    print("=" * 65)
```

**Result**

The program successfully created an exact binary copy of the selected evidence file. The original and forensic-copy files were compared using file size and SHA-256 hash values. Since both matched, the forensic copy was verified as an exact representation of the original evidence file.
