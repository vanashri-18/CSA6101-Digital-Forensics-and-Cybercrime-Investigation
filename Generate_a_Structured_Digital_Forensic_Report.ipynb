**Aim**

To automatically generate a structured digital forensic report from investigation findings using Python.

**Algorithm**

Import the required Python library.

Define the investigation details such as case ID, investigator name, evidence, findings, and conclusion.

Create a text file to store the forensic report.

Write the investigation details into the report in a structured format.

Save the report.

Display a success message after generating the report.

```python
from datetime import datetime

# Investigation Details
case_id = "CASE-2026-001"
investigator = "John Doe"
date = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

evidence = [
    "Laptop seized from suspect",
    "USB flash drive recovered",
    "Email header analyzed"
]

findings = [
    "Multiple failed login attempts detected.",
    "Suspicious executable file identified.",
    "Email spoofing indicators found.",
    "Port scanning activity observed."
]

conclusion = "Evidence indicates unauthorized access and malicious activities."

# Report file
report_file = "Digital_Forensic_Report.txt"

with open(report_file, "w") as file:
    file.write("DIGITAL FORENSIC REPORT\n")
    file.write("=" * 40 + "\n\n")
    file.write(f"Case ID       : {case_id}\n")
    file.write(f"Investigator  : {investigator}\n")
    file.write(f"Date          : {date}\n\n")

    file.write("Evidence Collected:\n")
    for item in evidence:
        file.write(f"- {item}\n")

    file.write("\nInvestigation Findings:\n")
    for item in findings:
        file.write(f"- {item}\n")

    file.write("\nConclusion:\n")
    file.write(conclusion)

print("Digital Forensic Report generated successfully.")
print("Report saved as:", report_file)

# Display report
with open(report_file, "r") as file:
    print("\n")
    print(file.read())
```

**Result**

The Python program successfully generated a structured digital forensic report containing the case details, evidence collected, investigation findings, and conclusion. The report was saved as a text file and displayed successfully.
