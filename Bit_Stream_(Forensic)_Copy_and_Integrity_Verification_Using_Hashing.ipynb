**Aim**

To create a bit-stream (forensic) copy of a file and verify its integrity using the SHA-256 hashing algorithm in Python.

**Algorithm**

Import the required Python libraries (shutil and hashlib).

Upload or select the original file.

Create a bit-stream (byte-by-byte) copy of the file using shutil.copyfile().

Calculate the SHA-256 hash value of the original
file.

Calculate the SHA-256 hash value of the copied file.

Compare both hash values.

If both hashes are identical, confirm that the forensic copy is authentic and its integrity is preserved.

Otherwise, report that the file has been altered or corrupted.

```python
import shutil
import hashlib
from google.colab import files

# Function to calculate SHA-256 hash
def calculate_hash(file_path):
    sha256 = hashlib.sha256()
    with open(file_path, "rb") as f:
        while True:
            data = f.read(4096)
            if not data:
                break
            sha256.update(data)
    return sha256.hexdigest()

# Upload original file
print("Upload the original file:")
uploaded = files.upload()

original_file = list(uploaded.keys())[0]
copy_file = "forensic_copy_" + original_file

# Create forensic copy
shutil.copyfile(original_file, copy_file)

print("\nForensic copy created successfully!")

# Calculate hash values
original_hash = calculate_hash(original_file)
copy_hash = calculate_hash(copy_file)

print("\nOriginal Hash :", original_hash)
print("Copied Hash   :", copy_hash)

# Verify integrity
if original_hash == copy_hash:
    print("\nIntegrity Verification: SUCCESS")
    print("The forensic copy is identical to the original file.")
else:
    print("\nIntegrity Verification: FAILED")
    print("The copied file has been altered.")
```

**Result**

The bit-stream (forensic) copy of the selected file was successfully created. The SHA-256 hash values of the original and copied files were compared and found to be identical, confirming that the forensic copy is an exact duplicate and its integrity has been successfully verified.
