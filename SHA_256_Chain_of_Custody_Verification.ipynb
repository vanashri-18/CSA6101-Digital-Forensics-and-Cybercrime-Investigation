**Aim**

To verify the integrity of a digital evidence file by calculating its SHA-256 hash, comparing it with the originally recorded hash, and maintaining a simple chain-of-custody record.

**Algorithm**

1. Select Evidence:
Select the acquired digital evidence file and assign an evidence ID.

2. Record Hash:
Store the original SHA-256 hash in the evidence register.

3. Read Evidence:
Read the evidence file in binary mode.

4. Calculate Hash:
Generate the current SHA-256 hash using Python.

5. Compare Values:
Compare the original hash with the current hash.

6. Record Custody:
Store evidence ID, file name, acquisition time, original hash, verification time, and current hash.

7. Flag Changes:
If both hashes match, mark the evidence as VERIFIED. If they differ, flag it as POTENTIALLY ALTERED.

```python
import hashlib
from datetime import datetime
import os

# ---------------------------------------------------------
# SHA-256 DIGITAL EVIDENCE VERIFICATION
# ---------------------------------------------------------

EVIDENCE_ID = "EVID-001"
FILE_NAME = "digital_evidence.txt"

# ---------------------------------------------------------
# SHA-256 HASH FUNCTION
# ---------------------------------------------------------

def calculate_sha256(filename):

    sha256 = hashlib.sha256()

    with open(filename, "rb") as file:

        while True:
            data = file.read(4096)

            if not data:
                break

            sha256.update(data)

    return sha256.hexdigest()


# ---------------------------------------------------------
# CREATE TEST EVIDENCE FILE
# ---------------------------------------------------------

if not os.path.exists(FILE_NAME):

    with open(FILE_NAME, "w") as file:
        file.write("Digital forensic evidence file.\n")
        file.write("This file is used for integrity verification.\n")


# ---------------------------------------------------------
# ACQUISITION INFORMATION
# ---------------------------------------------------------

acquisition_time = datetime.now()

# Calculate original hash
original_hash = calculate_sha256(FILE_NAME)


# ---------------------------------------------------------
# DISPLAY ORIGINAL HASH
# ---------------------------------------------------------

print("=" * 70)
print("DIGITAL EVIDENCE ACQUISITION")
print("=" * 70)

print("Evidence ID     :", EVIDENCE_ID)
print("File Name       :", FILE_NAME)
print("Acquisition Time:",
      acquisition_time.strftime("%Y-%m-%d %H:%M:%S"))

print("Original SHA-256:")
print(original_hash)


# ---------------------------------------------------------
# VERIFICATION
# ---------------------------------------------------------

verification_time = datetime.now()

current_hash = calculate_sha256(FILE_NAME)


# ---------------------------------------------------------
# COMPARE HASHES
# ---------------------------------------------------------

print("\n" + "=" * 70)
print("HASH VERIFICATION")
print("=" * 70)

print("Verification Time:",
      verification_time.strftime("%Y-%m-%d %H:%M:%S"))

print("\nOriginal Hash:")
print(original_hash)

print("\nCurrent Hash:")
print(current_hash)


if original_hash == current_hash:

    status = "VERIFIED - EVIDENCE INTEGRITY MAINTAINED"

else:

    status = "POTENTIALLY ALTERED - HASH MISMATCH DETECTED"


print("\nVerification Status:", status)


# ---------------------------------------------------------
# CHAIN OF CUSTODY RECORD
# ---------------------------------------------------------

print("\n" + "=" * 70)
print("CHAIN-OF-CUSTODY RECORD")
print("=" * 70)

print("Evidence ID     :", EVIDENCE_ID)
print("File Name       :", FILE_NAME)
print("Acquisition Time:",
      acquisition_time.strftime("%Y-%m-%d %H:%M:%S"))
print("Original Hash   :", original_hash)
print("Verification Time:",
      verification_time.strftime("%Y-%m-%d %H:%M:%S"))
print("Current Hash    :", current_hash)
print("Status          :", status)


# ---------------------------------------------------------
# FORENSIC INTERPRETATION
# ---------------------------------------------------------

print("\n" + "=" * 70)
print("FORENSIC INTERPRETATION")
print("=" * 70)

if original_hash == current_hash:

    print("Observed Fact:")
    print("The original and current SHA-256 values are identical.")

    print("\nInterpretation:")
    print("No change to the file contents was detected by")
    print("the SHA-256 integrity check.")

else:

    print("Observed Fact:")
    print("The original and current SHA-256 values are different.")

    print("\nInterpretation:")
    print("The evidence file may have been altered after acquisition.")
    print("Further forensic examination is required.")

print("\n" + "=" * 70)
print("VERIFICATION COMPLETE")
print("=" * 70)
```

**Result**

The digital evidence file was successfully verified using SHA-256 hashing. The original and current hash values were compared and found to be identical, confirming that no change to the file contents was detected. The chain-of-custody record was also successfully maintained with the evidence ID, file name, acquisition time, original hash, verification time, current hash, and verification status.
