```python
# Digital Forensic Case Reporting System

def get_input(message):
    return input(message).strip()


print("DIGITAL FORENSIC CASE REPORTING SYSTEM")
print("-" * 45)

# Case information
case_id = get_input("Enter Case ID: ")
investigator = get_input("Enter Investigator Name: ")

# Evidence information
evidence_description = get_input("Enter Evidence Description: ")
acquisition_details = get_input("Enter Acquisition Details: ")
file_metadata = get_input("Enter File Metadata: ")
hash_value = get_input("Enter Hash Value: ")

# Validation of essential evidence information
if not evidence_description or not hash_value:
    print("\nERROR: Evidence description and hash value are required!")
else:
    # Investigation details
    examination = get_input("Enter Examination Activities: ")
    timeline = get_input("Enter Timeline Events: ")

    # Investigator analysis
    findings = get_input("Enter Findings/Observations: ")
    conclusions = get_input("Enter Conclusions: ")
    recommendations = get_input("Enter Recommendations: ")

    # Generate forensic report
    print("\n")
    print("=" * 60)
    print("DIGITAL FORENSIC INVESTIGATION REPORT")
    print("=" * 60)

    print("\nCASE INFORMATION")
    print("Case ID:", case_id)
    print("Investigator:", investigator)

    print("\nFACTUAL EVIDENCE")
    print("Evidence Description:", evidence_description)
    print("Acquisition Details:", acquisition_details)
    print("File Metadata:", file_metadata)
    print("Hash Value:", hash_value)

    print("\nEXAMINATION ACTIVITIES")
    print(examination)

    print("\nTIMELINE EVENTS")
    print(timeline)

    print("\nINVESTIGATOR FINDINGS / OBSERVATIONS")
    print(findings)

    print("\nCONCLUSIONS")
    print(conclusions)

    print("\nRECOMMENDATIONS")
    print(recommendations)

    print("\n" + "=" * 60)
    print("REPORT GENERATED SUCCESSFULLY")
    print("=" * 60)
```
