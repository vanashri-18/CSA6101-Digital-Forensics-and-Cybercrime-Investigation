**Aim**

To write a Python program to analyze a simulated Windows USB activity log and identify unauthorized USB device connections based on device ID, vendor ID, and authorization status.


**Algorithm**

Import the required Python libraries.

Create a simulated Windows USB activity log.

Load the log into a Pandas DataFrame.

Define the list of authorized USB device IDs.

Compare each connected USB device with the authorized device list.

Mark devices not present in the authorized list as unauthorized.

Display all unauthorized USB connections.

Generate security alerts for unauthorized devices.

Display the final analysis summary.

```python
import pandas as pd
from io import StringIO

# ==========================================================
# Simulated Windows USB Activity Log
# ==========================================================

usb_log = """Timestamp,Computer,User,DeviceName,DeviceID,VendorID,Action
2026-08-18 09:10:15,PC-01,Admin,Kingston DataTraveler,USB001,0951,Connected
2026-08-18 09:25:30,PC-01,User1,SanDisk Cruzer,USB002,0781,Connected
2026-08-18 10:05:45,PC-01,User2,Unknown USB Device,USB003,FFFF,Connected
2026-08-18 10:20:10,PC-01,Admin,Kingston DataTraveler,USB001,0951,Disconnected
2026-08-18 10:35:25,PC-01,User3,External Hard Drive,USB004,1058,Connected
2026-08-18 11:10:40,PC-01,User1,Unknown USB Device,USB005,1234,Connected
2026-08-18 11:25:55,PC-01,User2,SanDisk Cruzer,USB002,0781,Disconnected
"""

# ==========================================================
# Step 1: Load USB log into DataFrame
# ==========================================================

df = pd.read_csv(StringIO(usb_log))

print("========== WINDOWS USB ACTIVITY LOG ==========\n")
print(df.to_string(index=False))


# ==========================================================
# Step 2: Define Authorized USB Devices
# ==========================================================

authorized_devices = [
    "USB001",
    "USB002",
    "USB004"
]


# ==========================================================
# Step 3: Filter USB Connection Events
# ==========================================================

connections = df[df["Action"] == "Connected"].copy()

print("\n========== USB CONNECTION EVENTS ==========\n")
print(connections.to_string(index=False))


# ==========================================================
# Step 4: Identify Unauthorized Devices
# ==========================================================

connections["Status"] = connections["DeviceID"].apply(
    lambda x: "Authorized"
    if x in authorized_devices
    else "Unauthorized"
)

unauthorized = connections[
    connections["Status"] == "Unauthorized"
]


# ==========================================================
# Step 5: Display Unauthorized USB Devices
# ==========================================================

print("\n========== UNAUTHORIZED USB DEVICES ==========\n")

if unauthorized.empty:
    print("No unauthorized USB devices detected.")

else:
    print(
        unauthorized[
            [
                "Timestamp",
                "Computer",
                "User",
                "DeviceName",
                "DeviceID",
                "VendorID",
                "Status"
            ]
        ].to_string(index=False)
    )


# ==========================================================
# Step 6: Generate Security Alerts
# ==========================================================

print("\n========== SECURITY ALERTS ==========\n")

if unauthorized.empty:
    print("No security alerts generated.")

else:
    for _, row in unauthorized.iterrows():

        print(
            f"[ALERT] Unauthorized USB Device Detected | "
            f"User: {row['User']} | "
            f"Device: {row['DeviceName']} | "
            f"Device ID: {row['DeviceID']} | "
            f"Vendor ID: {row['VendorID']}"
        )


# ==========================================================
# Step 7: Analysis Summary
# ==========================================================

total_connections = len(connections)
authorized_count = len(
    connections[connections["Status"] == "Authorized"]
)
unauthorized_count = len(unauthorized)

print("\n========== ANALYSIS SUMMARY ==========\n")

print("Total USB Connections :", total_connections)
print("Authorized Devices    :", authorized_count)
print("Unauthorized Devices  :", unauthorized_count)

if unauthorized_count > 0:
    print("\nWARNING: Unauthorized USB activity detected!")

else:
    print("\nNo unauthorized USB activity detected.")

print("\nUSB activity analysis completed successfully.")
```

**Result**

The Python program successfully analyzed the simulated Windows USB activity log and identified unauthorized USB device connections by comparing connected device IDs with a predefined list of authorized devices. The program generated security alerts for unauthorized devices, which can help investigators detect potential data theft, malware introduction, or policy violations during digital forensic investigations.
