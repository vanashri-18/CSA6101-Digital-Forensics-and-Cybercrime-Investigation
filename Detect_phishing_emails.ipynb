Aim:

To develop a Python program that detects phishing emails by analyzing suspicious indicators present in the sender email, subject, and email body, and classifies the email as Safe, Suspicious, or Likely Phishing.

Alogirthm:

1.Start the program.

2.Define a list of suspicious keywords for the sender, subject, and email body.

3.Accept the sender email, subject, and body as input from the user.

4.Convert all inputs to lowercase for case-insensitive matching.
5.Initialize a phishing score to 0.
6.Compare the sender email with the suspicious sender keywords. If a match is found, increment the score.
7.Compare the subject with the suspicious subject keywords.
8.If a match is found, increment the score.
9.Compare the email body with the suspicious body keywords.
10.If a match is found, increment the score.

11.Calculate the total phishing score.

12.Classify the email:

Score ≥ 5 → Likely Phishing Email

Score 2–4 → Suspicious Email

Score < 2 → Safe Email

13.Display the phishing analysis and final result.

14.Stop the program.

```python
# Phishing Email Detection Program

sender_keywords=["unknown","fake","noreply",".xyz",".ru"]
subject_keywords=["urgent","verify","account","winner","lottery","free","claim","password"]
body_keywords=["click here","login","verify your account","bank","otp","password","credit card","update now","limited time","confirm"]

print("========== PHISHING EMAIL DETECTOR ==========")
sender=input("Enter Sender Email: ").lower()
subject=input("Enter Email Subject: ").lower()
body=input("Enter Email Body: ").lower()
score=0
print("\n----- ANALYSIS -----")
for k in sender_keywords:
    if k in sender:
        print("Suspicious Sender:",k); score+=1
for k in subject_keywords:
    if k in subject:
        print("Suspicious Subject:",k); score+=1
for k in body_keywords:
    if k in body:
        print("Suspicious Body:",k); score+=1
print("\nTotal Suspicious Indicators:",score)
if score>=5:
    print("Result: Likely Phishing Email")
elif score>=2:
    print("Result: Suspicious Email")
else:
    print("Result: Safe Email")

```
