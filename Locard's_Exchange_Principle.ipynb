AIM:

To demonstrate Locard's Exchange Principle by observing changes in file metadata (timestamps) after modifying a digital evidence file. This shows that every interaction with digital evidence leaves a trace.




Algorithm

1.Start the program.

2.Create a sample digital evidence file.

3.Display the file's creation, modification, and access timestamps.

4.Wait for a few seconds.

5.Modify the contents of the file.

6.Display the updated timestamps.

7.Compare the initial and updated timestamps.

8.Observe that the modification timestamp has changed, indicating that the file has been altered.

9.Display the conclusion that every interaction leaves a digital trace.

10.Stop the program.





```python
# ==========================================
# Locard's Exchange Principle Demonstration
# Using File Metadata (Timestamps)
# ==========================================

import os
import time
from datetime import datetime

# Step 1: Create a sample evidence file
filename = "evidence.txt"

with open(filename, "w") as file:
    file.write("Original Digital Evidence")

print("Evidence file created successfully.\n")

# Step 2: Function to display timestamps
def display_metadata(file_name):

    created = os.path.getctime(file_name)
    modified = os.path.getmtime(file_name)
    accessed = os.path.getatime(file_name)

    print("------ File Metadata ------")
    print("Created :", datetime.fromtimestamp(created))
    print("Modified:", datetime.fromtimestamp(modified))
    print("Accessed:", datetime.fromtimestamp(accessed))
    print("----------------------------\n")

print("Initial Metadata")
display_metadata(filename)

# Wait for 5 seconds
time.sleep(5)

# Step 3: Simulate evidence modification
with open(filename, "a") as file:
    file.write("\nUnauthorized Modification")

print("Evidence modified.\n")

print("Updated Metadata")
display_metadata(filename)

print("Conclusion:")
print("The modification timestamp changed after altering the file.")
print("This demonstrates Locard's Exchange Principle—every interaction leaves a trace.")
```
