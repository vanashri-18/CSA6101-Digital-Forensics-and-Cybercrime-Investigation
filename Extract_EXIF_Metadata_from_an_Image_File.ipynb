**Aim**

To extract and display EXIF (Exchangeable Image File Format) metadata embedded in an image file using Python.

**Algorithm**

Import the required Python libraries (Pillow).

Upload the image file.

Open the image using the Pillow library.

Extract the EXIF metadata from the image.

Convert EXIF tag IDs into readable tag names.

Display all available EXIF metadata.

If no EXIF metadata exists, display an appropriate message.

```python
# Install Pillow (if required)
!pip install pillow

from PIL import Image
from PIL.ExifTags import TAGS
from google.colab import files

# Upload image
print("Upload an image file:")
uploaded = files.upload()

filename = list(uploaded.keys())[0]

# Open image
image = Image.open(filename)

# Extract EXIF metadata
exif_data = image.getexif()

if exif_data:
    print("\nEXIF Metadata\n----------------------")
    for tag_id, value in exif_data.items():
        tag = TAGS.get(tag_id, tag_id)
        print(f"{tag}: {value}")
else:
    print("No EXIF metadata found in the image.")
```

**Result**

The Python program successfully extracted and displayed the EXIF metadata embedded in the image file. The metadata included details such as the camera make and model, date and time, image resolution, software used, and GPS information (if available).
