**Aim**

To develop a Python program that collects publicly available WHOIS information for an investigator-supplied domain and displays available registration details, registrar information, domain dates, name servers, and domain status. The program should safely handle privacy-protected or unavailable fields without assuming or generating incorrect information.

**Algorithm**

Accept a domain name from the investigator.

Validate and normalize the domain name.

Query the WHOIS service.

Extract available registrar information.

Extract registration, creation, update, and expiry dates.

Extract name servers.

Extract domain status.

Handle missing or privacy-protected information as Not Available / Privacy Protected.

Record the queried domain.

Record the collection timestamp.

Display the collected information in a structured format.

```python
# ==============================================
# WHOIS INFORMATION COLLECTION TOOL
# ==============================================

import whois
import re
from datetime import datetime

# ------------------------------------------------
# 1. Helper Function
# ------------------------------------------------

def safe_value(value):

    if value is None:
        return "Not Available / Privacy Protected"

    if isinstance(value, (list, tuple, set)):

        values = []

        for item in value:

            if item:
                values.append(str(item))

        if values:
            return ", ".join(values)

        return "Not Available / Privacy Protected"

    return str(value)

# ------------------------------------------------
# 2. Domain Input
# ------------------------------------------------

domain = input(
    "Enter domain name: "
).strip().lower()

# Remove http/https if entered
domain = re.sub(
    r"^https?://",
    "",
    domain
)

# Remove path
domain = domain.split("/")[0]

collection_time = datetime.now().strftime(
    "%Y-%m-%d %H:%M:%S"
)

print("\n" + "=" * 85)
print("                 WHOIS INFORMATION COLLECTION")
print("=" * 85)

print("Queried Domain :", domain)
print("Collection Time:", collection_time)

# ------------------------------------------------
# 3. WHOIS Query
# ------------------------------------------------

try:

    record = whois.whois(domain)

    print("\n" + "-" * 85)
    print("                    REGISTRATION DETAILS")
    print("-" * 85)

    print(
        "Domain Name       :",
        safe_value(record.domain_name)
    )

    print(
        "Registrar         :",
        safe_value(record.registrar)
    )

    print(
        "Registrant        :",
        safe_value(record.name)
    )

    print(
        "Organization      :",
        safe_value(record.org)
    )

    print(
        "Registrant Country:",
        safe_value(record.country)
    )

    # ------------------------------------------------
    # Dates
    # ------------------------------------------------

    print("\n" + "-" * 85)
    print("                       DOMAIN DATES")
    print("-" * 85)

    print(
        "Creation Date     :",
        safe_value(record.creation_date)
    )

    print(
        "Updated Date      :",
        safe_value(record.updated_date)
    )

    print(
        "Expiration Date   :",
        safe_value(record.expiration_date)
    )

    # ------------------------------------------------
    # Name Servers
    # ------------------------------------------------

    print("\n" + "-" * 85)
    print("                       NAME SERVERS")
    print("-" * 85)

    print(
        safe_value(record.name_servers)
    )

    # ------------------------------------------------
    # Domain Status
    # ------------------------------------------------

    print("\n" + "-" * 85)
    print("                       DOMAIN STATUS")
    print("-" * 85)

    print(
        safe_value(record.status)
    )

    # ------------------------------------------------
    # DNS Information
    # ------------------------------------------------

    print("\n" + "-" * 85)
    print("                    ADDITIONAL DETAILS")
    print("-" * 85)

    print(
        "WHOIS Server      :",
        safe_value(record.whois_server)
    )

    print(
        "Registrar URL     :",
        safe_value(record.registrar_url)
    )

    print("\nWHOIS collection completed.")

except Exception as error:

    print("\n" + "=" * 85)
    print("                    WHOIS QUERY RESULT")
    print("=" * 85)

    print(
        "Unable to obtain complete WHOIS information."
    )

    print(
        "Reason:",
        str(error)
    )

    print(
        "\nNo unavailable field has been assumed or fabricated."
    )

print("=" * 85)
```

**Result**

The Python program successfully accepts an investigator-supplied domain and attempts to collect publicly available WHOIS information. It displays the queried domain and collection time, along with available registrar details, registration dates, name servers, and domain status. Missing or privacy-protected fields are explicitly reported as “Not Available / Privacy Protected” rather than being guessed, making the output suitable for basic forensic/OSINT documentation.
