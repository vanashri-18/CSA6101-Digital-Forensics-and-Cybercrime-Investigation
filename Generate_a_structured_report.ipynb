Aim

To develop a Python program that generates a structured Digital Forensic Investigation Report based on the five-stage digital forensic investigation lifecycle: Identification, Preservation, Collection, Examination & Analysis, and Reporting & Presentation.

Algorithm

1.Start the program.

2.Import the required modules (datetime and google.colab.files).

3.Initialize the case details such as Case ID, Investigator Name, Case Title, and Current Date.

4.Create a structured report containing the five stages of the digital forensic investigation lifecycle:

       Identification

       Preservation

       Collection

       Examination & Analysis

       Reporting & Presentation

5.Display the report on the screen.

6.Save the report as a text file (Digital_Forensic_Report.txt).

7.Download the generated report file using Google Colab.

8.End the program.

```python
# ==========================================
# Digital Forensic Investigation Report
# Google Colab Version
# ==========================================

from datetime import datetime

# Report Details
case_id = "DF-2026-001"
investigator = "Vanashri S.R."
case_title = "Unauthorized Access Investigation"
date = datetime.now().strftime("%d-%m-%Y")

# Create Report
report = f"""
============================================================
          DIGITAL FORENSIC INVESTIGATION REPORT
============================================================

Case ID       : {case_id}
Investigator  : {investigator}
Date          : {date}
Case Title    : {case_title}

============================================================
FIVE-STAGE DIGITAL FORENSIC INVESTIGATION LIFECYCLE
============================================================

1. IDENTIFICATION
------------------------------------------------------------
Objective:
Identify the cyber incident and determine possible evidence.

Activities:
• Detect suspicious activity.
• Identify affected systems.
• Locate potential digital evidence.
• Define investigation scope.

============================================================

2. PRESERVATION
------------------------------------------------------------
Objective:
Preserve digital evidence without modification.

Activities:
• Isolate compromised systems.
• Create forensic disk images.
• Generate MD5/SHA-256 hashes.
• Maintain Chain of Custody.

============================================================

3. COLLECTION
------------------------------------------------------------
Objective:
Collect all relevant digital evidence.

Activities:
• Acquire disk images.
• Collect RAM dumps.
• Gather system and application logs.
• Copy network traffic captures.

============================================================

4. EXAMINATION & ANALYSIS
------------------------------------------------------------
Objective:
Analyze collected evidence to determine attack details.

Activities:
• Recover deleted files.
• Analyze browser history.
• Examine event logs.
• Identify malware artifacts.
• Reconstruct attack timeline.

============================================================

5. REPORTING & PRESENTATION
------------------------------------------------------------
Objective:
Document findings and present investigation results.

Activities:
• Prepare forensic report.
• Summarize findings.
• Include evidence details.
• Recommend preventive measures.

============================================================
INVESTIGATION SUMMARY
============================================================

✔ Incident Successfully Identified
✔ Evidence Preserved
✔ Evidence Collected
✔ Evidence Analyzed
✔ Final Report Prepared

Status : Investigation Completed Successfully

============================================================
"""

# Display Report
print(report)

# Save Report to a Text File
filename = "Digital_Forensic_Report.txt"

with open(filename, "w") as file:
    file.write(report)

print(f"\nReport saved successfully as '{filename}'")

# Download file in Google Colab
from google.colab import files
files.download(filename)
```
