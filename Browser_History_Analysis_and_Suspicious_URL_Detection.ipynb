**Aim**

To write a Python program to analyze browser history data, identify frequently visited websites, and detect potentially suspicious URLs based on predefined indicators.

**Algorithm**

Import the required Python libraries.

Create or upload a browser history CSV file containing URLs and visit counts.

Read the browser history data using Pandas.

Extract the domain name from each URL.

Count the number of visits for each website.

Identify the most frequently visited websites.

Check URLs for suspicious indicators such as:

IP addresses instead of domain names

@ symbols

Suspicious keywords

Excessive subdomains

Unusual URL structures

Display frequently visited websites.

Display URLs identified as suspicious.

Generate the final analysis result.

```python
import pandas as pd
import re
from urllib.parse import urlparse
from google.colab import files

# Upload browser history CSV
print("Upload browser_history.csv")
uploaded = files.upload()

# Read CSV file
file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)

print("\nBrowser History Data")
print("=" * 60)
print(df)

# Function to extract domain
def extract_domain(url):
    try:
        parsed = urlparse(url)
        return parsed.netloc.lower()
    except:
        return "Unknown"

# Extract domain
df["domain"] = df["url"].apply(extract_domain)

# ---------------------------------------------------------
# FREQUENTLY VISITED WEBSITES
# ---------------------------------------------------------

domain_visits = df.groupby("domain")["visit_count"].sum().sort_values(
    ascending=False
)

print("\n" + "=" * 60)
print("FREQUENTLY VISITED WEBSITES")
print("=" * 60)

print(domain_visits.head(10))


# ---------------------------------------------------------
# SUSPICIOUS URL DETECTION
# ---------------------------------------------------------

suspicious_keywords = [
    "login",
    "verify",
    "account",
    "password",
    "free-prize",
    "winner",
    "bank",
    "urgent",
    "claim"
]

def detect_suspicious(url):

    url_lower = url.lower()
    reasons = []

    # Check for IP address
    ip_pattern = r"https?://(\d{1,3}\.){3}\d{1,3}"

    if re.search(ip_pattern, url_lower):
        reasons.append("IP address used instead of domain")

    # Check for @ symbol
    if "@" in url:
        reasons.append("@ symbol in URL")

    # Check for suspicious keywords
    for keyword in suspicious_keywords:
        if keyword in url_lower:
            reasons.append("Suspicious keyword: " + keyword)

    # Check for unusually long URL
    if len(url) > 100:
        reasons.append("Unusually long URL")

    # Check for excessive subdomains
    domain = extract_domain(url)

    if domain.count(".") >= 4:
        reasons.append("Excessive subdomains")

    if reasons:
        return ", ".join(reasons)

    return "Normal"


# Apply suspicious URL detection
df["analysis"] = df["url"].apply(detect_suspicious)

# Display suspicious URLs
suspicious_df = df[df["analysis"] != "Normal"]

print("\n" + "=" * 60)
print("SUSPICIOUS URLS")
print("=" * 60)

if len(suspicious_df) > 0:

    for index, row in suspicious_df.iterrows():

        print("\n[!] Suspicious URL")
        print("URL:", row["url"])
        print("Visits:", row["visit_count"])
        print("Reason:", row["analysis"])

else:
    print("No suspicious URLs detected.")


# ---------------------------------------------------------
# FINAL RESULT
# ---------------------------------------------------------

print("\n" + "=" * 60)
print("FINAL RESULT")
print("=" * 60)

print("Total URLs analyzed:", len(df))
print("Unique websites:", df["domain"].nunique())
print("Suspicious URLs:", len(suspicious_df))

print("\nMost Frequently Visited Website:")

if len(domain_visits) > 0:
    print(domain_visits.index[0])
    print("Total visits:", domain_visits.iloc[0])

print("\nBrowser history analysis completed successfully.")
```

**Result**

The Python program successfully analyzes browser history data and identifies frequently visited websites based on visit counts. It also detects potentially suspicious URLs using indicators such as IP-based URLs, @ symbols, suspicious keywords, excessive subdomains, and unusually long URLs. This analysis can help forensic investigators identify potential phishing, malicious, or suspicious browsing activity.
