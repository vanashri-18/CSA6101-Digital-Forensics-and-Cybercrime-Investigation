```python
# Digital Forensic Evidence Inventory System

evidence_inventory = []

# Function to add evidence
def add_evidence():
    item = {
        "number": input("Enter Evidence Number: "),
        "description": input("Enter Description: "),
        "type": input("Enter Evidence Type: "),
        "source": input("Enter Source: "),
        "acquisition": input("Enter Acquisition Method: "),
        "storage": input("Enter Storage Location: "),
        "size": input("Enter File Size: "),
        "hash": input("Enter Hash Value: "),
        "status": input("Enter Integrity Status: ")
    }

    evidence_inventory.append(item)
    print("\nEvidence added successfully!\n")


# Function to display evidence
def display_evidence(items):
    if not items:
        print("\nNo matching evidence found.")
        return

    for item in items:
        print("\n" + "-" * 40)
        print("Evidence Number:", item["number"])
        print("Description:", item["description"])
        print("Evidence Type:", item["type"])
        print("Source:", item["source"])
        print("Acquisition Method:", item["acquisition"])
        print("Storage Location:", item["storage"])
        print("File Size:", item["size"])
        print("Hash Value:", item["hash"])
        print("Integrity Status:", item["status"])


# Main program
while True:
    print("\nDIGITAL FORENSIC EVIDENCE INVENTORY")
    print("1. Add Evidence")
    print("2. Display All Evidence")
    print("3. Search by Evidence Type")
    print("4. Search by Source")
    print("5. Search by Integrity Status")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_evidence()

    elif choice == "2":
        display_evidence(evidence_inventory)

    elif choice == "3":
        evidence_type = input("Enter Evidence Type: ").lower()
        results = [
            item for item in evidence_inventory
            if item["type"].lower() == evidence_type
        ]
        display_evidence(results)

    elif choice == "4":
        source = input("Enter Source: ").lower()
        results = [
            item for item in evidence_inventory
            if item["source"].lower() == source
        ]
        display_evidence(results)

    elif choice == "5":
        status = input("Enter Integrity Status: ").lower()
        results = [
            item for item in evidence_inventory
            if item["status"].lower() == status
        ]
        display_evidence(results)

    elif choice == "6":
        print("Program Closed.")
        break

    else:
        print("Invalid choice! Please try again.")Suspect
```
