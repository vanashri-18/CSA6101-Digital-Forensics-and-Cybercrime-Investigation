**Aim**

To write a Python program to analyze a simulated process execution log and identify suspicious processes based on suspicious process names, file locations, and unusual command-line arguments.

**Algorithm**

Import the required Python libraries.

Create a simulated process execution log containing process details.

Load the log data into a Pandas DataFrame.

Define a list of known suspicious process names and suspicious keywords.

Check each process for suspicious indicators such as:

Unusual process names.

Execution from temporary folders.

Suspicious file extensions or locations.

Suspicious command-line arguments.

Mark processes containing suspicious indicators.

Display all identified suspicious processes.

Generate an alert for each suspicious process.

```python
import pandas as pd
from io import StringIO

# ---------------------------------------------
# Simulated Process Execution Log
# ---------------------------------------------

log_data = """Timestamp,PID,ProcessName,User,FilePath,CommandLine
2026-08-18 10:01:10,1024,explorer.exe,User1,C:\\Windows\\explorer.exe,explorer.exe
2026-08-18 10:02:15,2048,chrome.exe,User1,C:\\Program Files\\Google\\Chrome\\Application\\chrome.exe,chrome.exe
2026-08-18 10:03:20,3050,notepad.exe,User1,C:\\Windows\\System32\\notepad.exe,notepad.exe notes.txt
2026-08-18 10:05:10,4050,mimikatz.exe,Admin,C:\\Users\\Admin\\Downloads\\mimikatz.exe,mimikatz.exe privilege::debug
2026-08-18 10:06:25,5050,powershell.exe,User2,C:\\Windows\\System32\\WindowsPowerShell\\v1.0\\powershell.exe,powershell.exe -enc SQBFAFgA
2026-08-18 10:07:40,6050,svchost.exe,System,C:\\Windows\\System32\\svchost.exe,svchost.exe -k netsvcs
2026-08-18 10:08:55,7050,unknown.exe,User3,C:\\Users\\User3\\AppData\\Local\\Temp\\unknown.exe,unknown.exe -silent
2026-08-18 10:10:05,8050,cmd.exe,User1,C:\\Windows\\System32\\cmd.exe,cmd.exe /c whoami
2026-08-18 10:11:20,9050,malware.exe,User4,C:\\Temp\\malware.exe,malware.exe --hidden
"""

# Load log data into a DataFrame
df = pd.read_csv(StringIO(log_data))

print("========== PROCESS EXECUTION LOG ==========\n")
print(df.to_string(index=False))


# ---------------------------------------------
# Define Suspicious Indicators
# ---------------------------------------------

suspicious_processes = [
    "mimikatz.exe",
    "malware.exe",
    "unknown.exe"
]

suspicious_keywords = [
    "-enc",
    "privilege::debug",
    "--hidden",
    "-silent"
]

suspicious_paths = [
    "\\Temp\\",
    "\\Downloads\\",
    "\\AppData\\"
]


# ---------------------------------------------
# Function to Detect Suspicious Processes
# ---------------------------------------------

def is_suspicious(row):

    process_name = row["ProcessName"].lower()
    command_line = row["CommandLine"].lower()
    file_path = row["FilePath"].lower()

    # Check suspicious process name
    if process_name in suspicious_processes:
        return True

    # Check suspicious command-line arguments
    for keyword in suspicious_keywords:
        if keyword.lower() in command_line:
            return True

    # Check suspicious execution locations
    for path in suspicious_paths:
        if path.lower() in file_path:
            return True

    return False


# ---------------------------------------------
# Identify Suspicious Processes
# ---------------------------------------------

df["Suspicious"] = df.apply(is_suspicious, axis=1)

suspicious_df = df[df["Suspicious"] == True]


# ---------------------------------------------
# Display Results
# ---------------------------------------------

print("\n========== SUSPICIOUS PROCESSES DETECTED ==========\n")

if suspicious_df.empty:
    print("No suspicious processes detected.")
else:
    print(
        suspicious_df[
            ["Timestamp", "PID", "ProcessName", "User",
             "FilePath", "CommandLine"]
        ].to_string(index=False)
    )


# ---------------------------------------------
# Generate Security Alerts
# ---------------------------------------------

print("\n========== SECURITY ALERTS ==========\n")

if suspicious_df.empty:
    print("No security alerts generated.")
else:
    for _, row in suspicious_df.iterrows():
        print(
            f"[ALERT] Suspicious Process Detected: "
            f"{row['ProcessName']} | "
            f"PID: {row['PID']} | "
            f"User: {row['User']}"
        )


# ---------------------------------------------
# Analysis Summary
# ---------------------------------------------

print("\n========== ANALYSIS SUMMARY ==========\n")

print("Total Processes Analyzed :", len(df))
print("Suspicious Processes     :", len(suspicious_df))
print("Normal Processes         :", len(df) - len(suspicious_df))

print("\nProcess analysis completed successfully.")
```

**Result**

The Python program successfully analyzed the simulated process execution log and identified suspicious processes based on process names, suspicious execution paths, and unusual command-line arguments. Processes such as mimikatz.exe, unknown.exe, malware.exe, and suspicious PowerShell execution were detected and flagged. This helps in identifying potential malicious activity during digital forensic investigation and incident response.
