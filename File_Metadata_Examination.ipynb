**Aim**

To develop a Python metadata examination tool that accepts a file or directory as input and extracts important file-system metadata, including file name, absolute path, size, creation time, modification time, and last access time, presenting the information in a structured format for forensic comparison.

**Algorithm**

Accept a file or directory path from the investigator.

Check whether the supplied path exists.

If the input is a file, analyze that file.

If the input is a directory, identify all files within it.

Obtain file-system metadata using Python's os.stat().

Extract the file name and absolute
path.

Determine the file size in bytes.

Extract creation, modification, and last-access timestamps.

Convert Unix timestamps into readable date-time values.

Display the metadata in a table for easy comparison.

Save the metadata report as a CSV file.

```python
# ==============================================
# FILE METADATA EXAMINATION TOOL
# ==============================================

import os
import pandas as pd
from datetime import datetime

print("=" * 80)
print("                 FILE METADATA EXAMINATION")
print("=" * 80)

# ------------------------------------------------
# 1. Get Input
# ------------------------------------------------

path = input(
    "\nEnter file or directory path: "
).strip()

# ------------------------------------------------
# 2. Check Path
# ------------------------------------------------

if not os.path.exists(path):

    print("\nERROR: File or directory does not exist.")

else:

    # Convert to absolute path
    path = os.path.abspath(path)

    files = []

    # ------------------------------------------------
    # 3. Identify Files
    # ------------------------------------------------

    if os.path.isfile(path):

        files.append(path)

    elif os.path.isdir(path):

        for root, directories, filenames in os.walk(path):

            for filename in filenames:

                files.append(
                    os.path.join(root, filename)
                )

    # ------------------------------------------------
    # 4. Extract Metadata
    # ------------------------------------------------

    records = []

    for filepath in files:

        try:

            information = os.stat(filepath)

            name = os.path.basename(filepath)

            absolute_path = os.path.abspath(filepath)

            size = information.st_size

            creation_time = datetime.fromtimestamp(
                information.st_ctime
            ).strftime("%Y-%m-%d %H:%M:%S")

            modification_time = datetime.fromtimestamp(
                information.st_mtime
            ).strftime("%Y-%m-%d %H:%M:%S")

            access_time = datetime.fromtimestamp(
                information.st_atime
            ).strftime("%Y-%m-%d %H:%M:%S")

            records.append({
                "File Name": name,
                "Absolute Path": absolute_path,
                "Size (Bytes)": size,
                "Creation Time": creation_time,
                "Modification Time": modification_time,
                "Last Access Time": access_time
            })

        except OSError as error:

            print(
                f"Could not examine {filepath}: {error}"
            )

    # ------------------------------------------------
    # 5. Display Results
    # ------------------------------------------------

    if records:

        report = pd.DataFrame(records)

        print("\n" + "=" * 80)
        print("                 METADATA REPORT")
        print("=" * 80)

        print(
            report.to_string(index=False)
        )

        # ------------------------------------------------
        # 6. Save Report
        # ------------------------------------------------

        output_file = "/content/file_metadata_report.csv"

        report.to_csv(
            output_file,
            index=False
        )

        print("\n" + "-" * 80)
        print("Files Examined :", len(report))
        print("Report Saved    :", output_file)
        print("-" * 80)

    else:

        print("\nNo files available for examination.")

print("\nMetadata examination completed.")
```

**Result**

The Python metadata examination tool successfully accepted a file or directory and extracted the available file-system metadata, including file name, absolute path, size, creation time, modification time, and last access time. The timestamps were converted into a readable date-time format and the results were organized into a structured table. The metadata was also exported to a CSV file, enabling investigators to conveniently compare multiple files during forensic examination.
