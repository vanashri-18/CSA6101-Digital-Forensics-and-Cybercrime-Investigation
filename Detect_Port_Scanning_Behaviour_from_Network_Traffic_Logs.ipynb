**Aim**

To analyze a simulated network traffic log and detect port-scanning behavior by identifying source IP addresses attempting to connect to multiple destination ports using Python.

**Algorithm**

Import the required Python libraries.

Upload the simulated network traffic log file.

Read the log file line by line.

Extract the source IP address and destination port.

Store the unique destination ports accessed by each source IP.

Compare the number of unique ports against a threshold (e.g., 5 ports).

Display IP addresses that exceed the threshold as potential port scanners.

```python
from google.colab import files
from collections import defaultdict
import re

# Upload network log file
print("Upload the network traffic log file:")
uploaded = files.upload()

filename = list(uploaded.keys())[0]

# Dictionary to store scanned ports
port_scan = defaultdict(set)

# Read log file
with open(filename, "r", encoding="utf-8", errors="ignore") as file:
    for line in file:
        # Example log:
        # SRC=192.168.1.10 DST=192.168.1.100 DPT=80
        src = re.search(r'SRC=([\d\.]+)', line)
        port = re.search(r'DPT=(\d+)', line)

        if src and port:
            ip = src.group(1)
            dport = port.group(1)
            port_scan[ip].add(dport)

# Threshold for port scanning
threshold = 5

print("\nPossible Port Scanners:\n")

found = False

for ip, ports in port_scan.items():
    if len(ports) >= threshold:
        print(f"Source IP: {ip}")
        print(f"Ports Scanned: {sorted(ports)}")
        print(f"Total Ports: {len(ports)}\n")
        found = True

if not found:
    print("No port-scanning behavior detected.")
```

**Result**

The Python program successfully analyzed the simulated network traffic log and detected potential port-scanning behavior by identifying source IP addresses that attempted connections to multiple unique destination ports beyond the specified threshold.
