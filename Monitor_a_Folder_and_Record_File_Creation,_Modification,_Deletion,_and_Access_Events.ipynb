**Aim**

To write a Python program that monitors a specified folder and records file creation, modification, deletion, and access events for digital forensic investigation.

**Algorithm**

Import the required Python libraries (watchdog, os, time, datetime).

Install the watchdog package if it is not already installed.

Specify the folder path to be monitored.

Create an event handler to detect file creation, modification, deletion, and movement events.

Record each event with the current date and time.

Continuously monitor the folder until the program is stopped.

Display the recorded events on the screen.

```python
# Install watchdog
!pip install watchdog

import os
import time
from datetime import datetime
from watchdog.observers import Observer
from watchdog.events import FileSystemEventHandler

# Create a folder to monitor
folder_to_monitor = "/content/MonitorFolder"
os.makedirs(folder_to_monitor, exist_ok=True)

print("Monitoring Folder:", folder_to_monitor)
print("Create, modify, or delete files in this folder.")
print("Press Stop (■) in Colab to end monitoring.\n")

# Event Handler
class FolderMonitor(FileSystemEventHandler):

    def log_event(self, event_type, file_path):
        print(f"[{datetime.now().strftime('%Y-%m-%d %H:%M:%S')}] {event_type}: {file_path}")

    def on_created(self, event):
        if not event.is_directory:
            self.log_event("File Created", event.src_path)

    def on_modified(self, event):
        if not event.is_directory:
            self.log_event("File Modified", event.src_path)

    def on_deleted(self, event):
        if not event.is_directory:
            self.log_event("File Deleted", event.src_path)

    def on_moved(self, event):
        if not event.is_directory:
            self.log_event("File Renamed/Moved", event.dest_path)

# Observer
event_handler = FolderMonitor()
observer = Observer()
observer.schedule(event_handler, folder_to_monitor, recursive=True)
observer.start()

try:
    while True:
        time.sleep(1)
except KeyboardInterrupt:
    observer.stop()

observer.join()
```

**Result**

The Python program successfully monitored the specified folder and recorded file creation, modification, deletion, and rename/move events with timestamps. This demonstrates how file activity can be tracked for digital forensic investigations.
