```python
from collections import defaultdict

# Sample authentication log data
logs = [
    {"timestamp": "10:00", "username": "admin", "ip": "192.168.1.10", "result": "FAIL"},
    {"timestamp": "10:01", "username": "admin", "ip": "192.168.1.10", "result": "FAIL"},
    {"timestamp": "10:02", "username": "admin", "ip": "192.168.1.10", "result": "FAIL"},

    {"timestamp": "10:05", "username": "user1", "ip": "10.0.0.5", "result": "FAIL"},
    {"timestamp": "10:06", "username": "user2", "ip": "10.0.0.5", "result": "FAIL"},
    {"timestamp": "10:07", "username": "user3", "ip": "10.0.0.5", "result": "FAIL"},
]

# Dictionaries for analysis
account_failures = defaultdict(int)
ip_accounts = defaultdict(set)

# Analyze failed authentication attempts
for log in logs:
    if log["result"] == "FAIL":
        username = log["username"]
        ip = log["ip"]

        account_failures[username] += 1
        ip_accounts[ip].add(username)

# Pattern 1: Repeated failures against one account
print("REPEATED FAILURES AGAINST ONE ACCOUNT")
for username, count in account_failures.items():
    if count >= 3:
        print(username, "->", count, "failed attempts")
        print("Possible Brute Force Attack\n")

# Pattern 2: Failures across multiple accounts from one IP
print("FAILURES AGAINST MULTIPLE ACCOUNTS")
for ip, users in ip_accounts.items():
    if len(users) >= 3:
        print(ip, "->", len(users), "different accounts:", users)
        print("Possible Password Spraying Attack\n")
```
